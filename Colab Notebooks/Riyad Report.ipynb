```python
import matplotlib.pyplot as plt
import numpy as np

# Set up the plot style
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(10, 6))

# Define time periods (Months 1 to 12)
months = np.arange(1, 13)
month_labels = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Define the Poverty Line (Constant threshold based on CBN method)
poverty_line = np.full(12, 5000)

# Simulate Stable Formal Income (Steady, slightly above poverty line)
formal_income = np.linspace(8000, 8500, 12)

# Simulate Volatile Informal Income (Fluctuating, drops during shocks)
informal_income = [7000, 7500, 7200, 6800, 3000, 4500, 6500, 7000, 7300, 2500, 4000, 6000]

# Plotting the lines
ax.plot(months, formal_income, color='#2ca02c', linewidth=2.5, label='Formal Sector Income (Stable)')
ax.plot(months, informal_income, color='#1f77b4', linewidth=2.5, marker='o', label='Informal Sector Income (Volatile)')
ax.axhline(y=5000, color='#d62728', linestyle='--', linewidth=2, label='Upper Poverty Line (CBN Method)')

# Highlight the "Vulnerability Traps"
ax.fill_between(months, informal_income, poverty_line,
                where=(np.array(informal_income) < poverty_line),
                interpolate=True, color='red', alpha=0.3, label='Extreme Poverty Gap')

# Annotate the Shocks
ax.annotate('Climate/Agricultural Shock\n(e.g., Cyclone/Flooding)', xy=(5, 3000), xytext=(3, 1500),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
            fontsize=10, fontweight='bold', ha='center')

ax.annotate('Sudden Health Crisis\n(Out-of-pocket expenses)', xy=(10, 2500), xytext=(11, 1000),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
            fontsize=10, fontweight='bold', ha='center')

# Formatting the chart
ax.set_xticks(months)
ax.set_xticklabels(month_labels)
ax.set_ylabel('Household Income / Expenditure (Theoretical Taka)', weight='bold')
ax.set_title('The Survivalist Trap: Income Volatility and Localized Shocks', weight='bold', fontsize=14, pad=15)
ax.legend(loc='upper right', frameon=True, shadow=True)

# Clean up borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('Figure_1_4_Income_Volatility.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the Survey Data (Chuadanga District, N=300 subset)
# Wages are in BDT. Women's wages are calculated to reflect a 28-35% deficit.
data = {
    'Informal Trade': [
        'Agricultural Labor',
        'Street Vending',
        'Transport (Van/Easy-bike)',
        'Cottage Industry',
        'Construction Day Labor'
    ],
    'Men_Wage_BDT': [500, 600, 800, 450, 700],
    'Women_Wage_BDT': [340, 420, 550, 324, 455]
}

df = pd.DataFrame(data)

# Calculate exact percentage deficit for data labels
df['Deficit_Pct'] = ((df['Men_Wage_BDT'] - df['Women_Wage_BDT']) / df['Men_Wage_BDT'] * 100).round(1)

# 2. Setup the Plot Figure
plt.rcParams['font.family'] = 'serif' # Academic serif font (e.g., Times New Roman style)
fig, ax = plt.subplots(figsize=(10, 6))

# Set bar width and positions
bar_width = 0.35
index = np.arange(len(df['Informal Trade']))

# 3. Create the Clustered Bars
bars_men = ax.bar(index, df['Men_Wage_BDT'], bar_width,
                  label='Men', color='#2c3e50', edgecolor='black')

bars_women = ax.bar(index + bar_width, df['Women_Wage_BDT'], bar_width,
                    label='Women', color='#e74c3c', edgecolor='black')

# 4. Add Data Labels (Values and Deficit Percentages)
def add_labels(bars, is_women=False):
    for i, bar in enumerate(bars):
        height = bar.get_height()
        # Add the wage value
        ax.annotate(f'৳{int(height)}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=10)

        # Add the deficit percentage above women's bars
        if is_women:
            deficit = df['Deficit_Pct'].iloc[i]
            ax.annotate(f'(-{deficit}%)',
                        xy=(bar.get_x() + bar.get_width() / 2, height),
                        xytext=(0, 15),  # Higher offset for the percentage
                        textcoords="offset points",
                        ha='center', va='bottom', fontsize=9, color='darkred', fontweight='bold')

add_labels(bars_men)
add_labels(bars_women, is_women=True)

# 5. Formatting and Academic Styling
ax.set_xlabel('Informal Trade Sector (Chuadanga)', fontsize=12, fontweight='bold', labelpad=10)
ax.set_ylabel('Average Daily Wage (BDT)', fontsize=12, fontweight='bold', labelpad=10)
ax.set_title('Figure 5.3: Gender Wage Disparities in the Chuadanga Informal Economy (N=300)',
             fontsize=14, fontweight='bold', pad=20)

ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(df['Informal Trade'], rotation=15, ha='right', fontsize=11)
ax.legend(loc='upper left', frameon=True, fontsize=11)

# Clean up grid and borders
ax.yaxis.grid(True, linestyle='--', alpha=0.7)
ax.set_axisbelow(True)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 6. Adjust layout and Save High-Resolution Image for PDF/Word
plt.tight_layout()
plt.savefig('Figure_5_3_Gender_Wage_Gap_Chuadanga.png', dpi=300, bbox_inches='tight')

# Display the plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the Survey Data (Percentages must sum to 100 for each trade)
data = {
    'Informal Trade': [
        'Agricultural Labor',
        'Street Vending',
        'Transport (Van/Easy-bike)',
        'Cottage Industry',
        'Construction'
    ],
    'No Debt (%)': [15, 20, 5, 25, 18],
    'Single Loan (%)': [45, 50, 25, 55, 42],
    'Multiple Loans (%)': [40, 30, 70, 20, 40] # High debt heavily concentrated in Transport
}

df = pd.DataFrame(data)

# 2. Setup the Plot Figure
plt.rcParams['font.family'] = 'serif'
fig, ax = plt.subplots(figsize=(10, 6.5))

# Define the academic color palette (Slate Blue, Neutral Grey, Warning Red)
colors = ['#2c3e50', '#bdc3c7', '#c0392b']
categories = ['No Debt (%)', 'Single Loan (%)', 'Multiple Loans (%)']

# 3. Create the Stacked Bars
bottom_arr = np.zeros(len(df)) # Keep track of the starting height for each stack

for i, col in enumerate(categories):
    # Plot the bar segment
    bars = ax.bar(df['Informal Trade'], df[col], bottom=bottom_arr,
                  label=col.replace(' (%)', ''), color=colors[i], edgecolor='white', width=0.6)

    # Add percentage labels inside the bars
    for bar in bars:
        height = bar.get_height()
        if height > 5: # Only show label if the section is large enough
            ax.annotate(f'{int(height)}%',
                        xy=(bar.get_x() + bar.get_width() / 2, bar.get_y() + height / 2),
                        ha='center', va='center', color='white', fontsize=11, fontweight='bold')

    # Update the bottom array for the next stack
    bottom_arr += df[col].values

# 4. Formatting and Academic Styling
ax.set_ylabel('Percentage of Households', fontsize=12, fontweight='bold', labelpad=10)
ax.set_title('Figure 6.1: Microcredit Debt Profiles by Informal Trade (N=300)',
             fontsize=14, fontweight='bold', pad=20)

ax.set_xticklabels(df['Informal Trade'], rotation=15, ha='right', fontsize=11)

# Format the Y-axis as percentages
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.0f}%'.format(y)))

# Legend configuration
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15),
          ncol=3, frameon=False, fontsize=11)

# Clean up layout
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#dddddd')
ax.spines['bottom'].set_color('#dddddd')
ax.tick_params(axis='x', bottom=False)
ax.tick_params(axis='y', left=False)

# 5. Save High-Resolution Image for PDF/Word
plt.tight_layout()
plt.savefig('Figure_6_1_Debt_Reliance.png', dpi=300, bbox_inches='tight')

# Display the plot
plt.show()
```

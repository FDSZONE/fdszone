```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def generate_aesthetic_helix():
    fig, ax = plt.subplots(figsize=(9, 7))
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off') # Hide axes for a clean infographic look

    # Modern Material Design Color Palette
    core_color = '#1A365D'   # Deep Navy
    colors = {
        'Gov': '#2B6CB0',    # Tech Blue
        'Private': '#319795',# Teal
        'NGO': '#DD6B20',    # Warm Orange
        'Academia': '#805AD5'# Purple
    }

    # Define Node Positions (Center, Top, Right, Bottom, Left)
    pos = {
        'Core': (0.5, 0.5),
        'Gov': (0.5, 0.85),
        'Private': (0.85, 0.5),
        'Academia': (0.5, 0.15),
        'NGO': (0.15, 0.5)
    }

    # 1. Draw Connecting Lines (Edges) first so they go behind the circles
    # Draw cross lines to center
    for key in ['Gov', 'Private', 'Academia', 'NGO']:
        ax.plot([pos['Core'][0], pos[key][0]], [pos['Core'][1], pos[key][1]],
                color='#CBD5E0', linewidth=3, zorder=1)

    # Draw outer boundary lines connecting the four pillars
    outer_keys = ['Gov', 'Private', 'Academia', 'NGO', 'Gov']
    for i in range(4):
        ax.plot([pos[outer_keys[i]][0], pos[outer_keys[i+1]][0]],
                [pos[outer_keys[i]][1], pos[outer_keys[i+1]][1]],
                color='#E2E8F0', linewidth=2, linestyle='--', zorder=1)

    # 2. Draw Nodes (Circles)
    def add_node(x, y, radius, color, text, font_color='white', font_size=11):
        circle = patches.Circle((x, y), radius, facecolor=color, edgecolor='white', linewidth=3, zorder=2)
        ax.add_patch(circle)
        ax.text(x, y, text, ha='center', va='center', color=font_color,
                fontsize=font_size, fontweight='bold', zorder=3)

    # Outer Nodes
    add_node(*pos['Gov'], 0.12, colors['Gov'], "Local Govt.\n(Union Parishad)")
    add_node(*pos['Private'], 0.12, colors['Private'], "Private Sector\n(Telecoms)")
    add_node(*pos['Academia'], 0.12, colors['Academia'], "Academia\n(Institutions)")
    add_node(*pos['NGO'], 0.12, colors['NGO'], "NGOs\n(e.g., BRAC)")

    # Core Node (Slightly larger)
    add_node(*pos['Core'], 0.16, core_color, "Localized Tech\nEcosystem\n(Chuadanga)", font_size=12)

    # Title
    plt.title("Figure 8.1: The Quadruple Helix Partnership Model",
              fontsize=16, fontweight='bold', color='#2D3748', pad=20)

    # Save and Show
    plt.tight_layout()
    plt.savefig('Figure_8_1_Aesthetic_Helix.png', dpi=300, bbox_inches='tight', transparent=False, facecolor='white')
    plt.show()

generate_aesthetic_helix()
```

```python
import matplotlib.pyplot as plt

def generate_aesthetic_roadmap():
    fig, ax = plt.subplots(figsize=(10, 5))
    fig.patch.set_facecolor('white') # Ensure background is pure white
    ax.set_facecolor('white')

    # Roadmap Data
    phases = [
        'Phase 4: Ecosystem Autonomy\n(Self-sustaining hubs)',
        'Phase 3: E-Agriculture Integration\n(IoT & Supply Chain)',
        'Phase 2: Literacy & Access\n(Device finance & Training)',
        'Phase 1: Infrastructural Stabilization\n(Solar grids & Broadband)'
    ]
    starts = [48, 24, 12, 0]
    durations = [12, 24, 12, 12]

    # Modern Gradient Color Palette (Light Blue to Dark Blue)
    colors = ['#63B3ED', '#3182CE', '#2B6CB0', '#2A4365']

    # Create the floating bars
    bars = ax.barh(phases, durations, left=starts, color=colors, height=0.55, edgecolor='none')

    # Add inner text labels (Months)
    for bar, duration, start in zip(bars, durations, starts):
        text_x = start + (duration / 2)
        text_y = bar.get_y() + bar.get_height() / 2
        ax.text(text_x, text_y, f'{duration} Months', ha='center', va='center',
                color='white', fontweight='bold', fontsize=11)

    # Minimalist Formatting (Removing clutter)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.spines['bottom'].set_color('#A0AEC0')

    # Formatting axes
    ax.set_xlabel('Project Timeline (Months)', fontsize=12, fontweight='bold', color='#4A5568')
    ax.set_xticks(range(0, 66, 6))
    ax.tick_params(axis='x', colors='#4A5568', labelsize=10)
    ax.tick_params(axis='y', left=False, colors='#2D3748', labelsize=11, pad=10)

    # Subtle vertical grid lines
    ax.xaxis.grid(True, linestyle='-', color='#EDF2F7', alpha=1, zorder=0)
    ax.set_axisbelow(True) # Ensure grid is behind the bars

    # Title
    plt.title('Figure 8.2: 60-Month Implementation Roadmap',
              fontsize=16, fontweight='bold', color='#2D3748', pad=20, loc='left')

    # Save and Show
    plt.tight_layout()
    plt.savefig('Figure_8_2_Aesthetic_Roadmap.png', dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()

generate_aesthetic_roadmap()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set global style for academic reporting
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12, 'figure.dpi': 300})

def generate_figure_5_2():
    """Figure 5.2: TCV Reduction Bar Chart"""
    labels = ['Time Required', 'Financial Cost', 'Physical Visits']
    traditional = [100, 100, 100] # Baseline 100%
    udc_impact = [15, 37, 20]     # Reduced to these percentages (85% and 63% reductions)

    x = np.arange(len(labels))
    width = 0.35

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.bar(x - width/2, traditional, width, label='Traditional Upazila HQ', color='gray')
    ax.bar(x + width/2, udc_impact, width, label='Union Digital Center (UDC)', color='teal')

    ax.set_ylabel('Percentage (%) of Baseline')
    ax.set_title('Figure 5.2: TCV Reduction via Union Digital Centers')
    ax.set_xticks(x)
    ax.set_xticklabels(labels)
    ax.legend()
    plt.tight_layout()
    plt.savefig('Figure_5_2_TCV_Reduction.png')
    plt.show()

def generate_figure_6_1():
    """Figure 6.1: Household Income Comparison"""
    groups = ['Non-Beneficiary Baseline', 'ICT & MFS Beneficiary']
    income_index = [100, 128] # 28% increase

    fig, ax = plt.subplots(figsize=(6, 5))
    bars = ax.bar(groups, income_index, color=['lightcoral', 'seagreen'])

    ax.set_ylabel('Relative Income Index')
    ax.set_title('Figure 6.1: Impact of Digital Interventions on Income')
    ax.set_ylim(0, 150)

    # Add percentage label on top of bars
    for bar in bars:
        yval = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, yval + 2, f'{yval}%', ha='center', va='bottom', fontweight='bold')

    plt.tight_layout()
    plt.savefig('Figure_6_1_Income_Comparison.png')
    plt.show()

def generate_figure_6_4():
    """Figure 6.4: Correlation Scatter Plot (Simulated Data)"""
    np.random.seed(42)
    # Simulate 422 households (your sample size)
    ict_access_score = np.random.normal(50, 15, 422)
    # Create a positive correlation for income growth
    income_growth = (ict_access_score * 0.4) + np.random.normal(10, 5, 422)

    fig, ax = plt.subplots(figsize=(8, 6))
    sns.regplot(x=ict_access_score, y=income_growth, ax=ax, scatter_kws={'alpha':0.5, 'color':'steelblue'}, line_kws={'color':'darkred'})

    ax.set_xlabel('ICT Access Index Score (0-100)')
    ax.set_ylabel('Household Income Growth Rate (%)')
    ax.set_title('Figure 6.4: Correlation between ICT Access and Income Growth')
    plt.tight_layout()
    plt.savefig('Figure_6_4_Correlation.png')
    plt.show()

# Execute functions to generate and save charts
generate_figure_5_2()
generate_figure_6_1()
generate_figure_6_4()
```

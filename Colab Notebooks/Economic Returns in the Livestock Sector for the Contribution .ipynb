```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import seaborn as sns

# Set advanced aesthetic style
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']

# Color Palettes
color_primary = "#1E3A8A"   # Deep Blue
color_secondary = "#0D9488" # Teal
color_accent = "#F26419"    # Soft Coral/Orange
color_bg = "#F9FAFB"        # Off-white background

def create_flowchart():
    """Generates Figure 1: Intervention Logic Flowchart"""
    fig, ax = plt.subplots(figsize=(10, 6), facecolor='white')
    ax.axis('off')

    # Define box properties
    box_style = "round,pad=0.5"

    # Nodes (Coordinates and Text)
    nodes = {
        "Constraints": {"pos": (0.1, 0.7), "text": "Baseline Constraints\n- Traditional Methods (97%)\n- Lack of GAP (91%)\n- Middlemen Dependence", "color": "#EF4444"},
        "Interventions": {"pos": (0.5, 0.7), "text": "SDI-RMTP Interventions\n- Capacity Building\n- Technology Adoption\n- Market Linkages", "color": color_primary},
        "Outcomes": {"pos": (0.9, 0.7), "text": "Economic Outcomes\n- +40% Income Increase\n- Higher Profit Margins\n- Job Generation", "color": color_secondary},
        "Empowerment": {"pos": (0.5, 0.3), "text": "Gender Empowerment\n- 92% Female Participation\n- 78% Decision Making", "color": color_accent}
    }

    # Draw Nodes
    for key, data in nodes.items():
        bbox_props = dict(boxstyle=box_style, fc=data["color"], ec="none", alpha=0.9)
        ax.text(data["pos"][0], data["pos"][1], data["text"], ha="center", va="center",
                size=11, color="white", weight="bold", bbox=bbox_props)

    # Draw Arrows
    arrow_props = dict(arrowstyle="-|>", color="gray", lw=2)
    ax.annotate("", xy=(0.35, 0.7), xytext=(0.25, 0.7), arrowprops=arrow_props) # Const -> Interv
    ax.annotate("", xy=(0.75, 0.7), xytext=(0.65, 0.7), arrowprops=arrow_props) # Interv -> Out
    ax.annotate("", xy=(0.5, 0.45), xytext=(0.5, 0.58), arrowprops=arrow_props) # Interv -> Empow
    ax.annotate("", xy=(0.85, 0.6), xytext=(0.65, 0.35), arrowprops=dict(arrowstyle="-|>", color="gray", lw=2, connectionstyle="arc3,rad=-0.2")) # Empow -> Out

    plt.title("Figure 1: Project Intervention Logic Model", fontsize=14, weight="bold", pad=20, color="#333333")
    plt.tight_layout()
    plt.savefig("Figure_1_Flowchart.png", dpi=300, bbox_inches='tight')
    plt.show()

def create_bar_chart():
    """Generates Figure 2: Profit Margins Bar Graph"""
    fig, ax = plt.subplots(figsize=(8, 5), facecolor='white')

    categories = ['Beef Fattening', 'Processed Goods', 'Milk Production']
    baseline = [24, 19, 11]
    # Milk production endline isn't explicitly stated in the text, assuming conservative +20% target (31%) for visual completeness based on text context.
    endline = [36, 33, 31]

    x = np.arange(len(categories))
    width = 0.35

    rects1 = ax.bar(x - width/2, baseline, width, label='Baseline', color="#94A3B8")
    rects2 = ax.bar(x + width/2, endline, width, label='Endline', color=color_secondary)

    # Customization
    ax.set_ylabel('Profit Margin (%)', fontsize=12, weight='bold')
    ax.set_title('Figure 2: Profit Margins by Enterprise Type (Baseline vs. Endline)', fontsize=14, weight='bold', pad=20)
    ax.set_xticks(x)
    ax.set_xticklabels(categories, fontsize=11)
    ax.legend(frameon=False, fontsize=11)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    # Add value labels
    def autolabel(rects):
        for rect in rects:
            height = rect.get_height()
            ax.annotate(f'{height}%',
                        xy=(rect.get_x() + rect.get_width() / 2, height),
                        xytext=(0, 3), textcoords="offset points",
                        ha='center', va='bottom', fontsize=10, weight='bold')

    autolabel(rects1)
    autolabel(rects2)

    plt.tight_layout()
    plt.savefig("Figure_2_BarGraph.png", dpi=300, bbox_inches='tight')
    plt.show()

def create_donut_chart():
    """Generates Figure 3: Demographics Donut Chart"""
    fig, ax = plt.subplots(figsize=(7, 6), facecolor='white')

    labels = ['Women', 'Men']
    sizes = [92, 8]
    colors = [color_accent, '#CBD5E1']
    explode = (0.05, 0)

    wedges, texts, autotexts = ax.pie(sizes, explode=explode, labels=labels, colors=colors,
                                      autopct='%1.1f%%', startangle=90, pctdistance=0.85,
                                      textprops={'fontsize': 12, 'weight': 'bold'})

    # Draw white circle in center to make it a donut
    centre_circle = plt.Circle((0,0),0.70,fc='white')
    fig.gca().add_artist(centre_circle)

    # Add central text
    plt.text(0, 0.1, '92%', ha='center', va='center', fontsize=28, weight='bold', color=color_accent)
    plt.text(0, -0.15, 'Female\nBeneficiaries', ha='center', va='center', fontsize=11, color='#475569')

    plt.title('Figure 3: Beneficiary Demographic Distribution', fontsize=14, weight='bold', pad=20)

    # Equal aspect ratio ensures that pie is drawn as a circle
    ax.axis('equal')
    plt.tight_layout()
    plt.savefig("Figure_3_DonutChart.png", dpi=300, bbox_inches='tight')
    plt.show()

# Execute functions to generate plots
if __name__ == "__main__":
    create_flowchart()
    create_bar_chart()
    create_donut_chart()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import graphviz
import os

# Set advanced aesthetic style for standard charts
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']

def generate_figure_1():
    """Generates a grouped bar chart for Income and Nutrition (MDD-W) comparisons."""
    labels = ['MDD-W Threshold Achieved (%)', 'Income Growth vs Baseline (%)']
    project_data = [89.4, 40.0] # 40% higher income growth represented here
    control_data = [72.06, 0.0] # Control serves as the 0 baseline for income difference

    x = np.arange(len(labels))
    width = 0.35

    fig, ax = plt.subplots(figsize=(8, 6))
    rects1 = ax.bar(x - width/2, project_data, width, label='Project Group (SDI/RMTP)', color='#2b5c8f', edgecolor='black')
    rects2 = ax.bar(x + width/2, control_data, width, label='Control Group', color='#d3d3d3', edgecolor='black')

    ax.set_ylabel('Percentage (%)', fontweight='bold')
    ax.set_title('Figure 1: Project Impact on Nutrition and Relative Income', fontweight='bold', pad=20)
    ax.set_xticks(x)
    ax.set_xticklabels(labels, fontweight='bold')
    ax.legend(frameon=True, shadow=True)

    # Add value labels on top of bars
    ax.bar_label(rects1, padding=3, fmt='%.1f%%')
    ax.bar_label(rects2, padding=3, fmt='%.1f%%')

    sns.despine()
    plt.tight_layout()
    plt.savefig('Figure_1_Impact_Comparison.png', dpi=300)
    plt.close()
    print("Figure 1 generated successfully.")

def generate_figure_2():
    """Generates a horizontal bar chart for Empowerment and Sustainability metrics."""
    metrics = [
        'Continued Tech Adoption',
        'Involvement in Decision-Making',
        'Plans to Expand Enterprise',
        'Access to Livestock Loans'
    ]
    percentages = [82, 78, 74, 70]

    fig, ax = plt.subplots(figsize=(9, 5))

    # Custom color palette mapping to values
    colors = sns.color_palette("viridis", len(percentages))
    bars = ax.barh(metrics, percentages, color=colors, edgecolor='black', height=0.6)

    ax.set_xlabel('Percentage of Participants (%)', fontweight='bold')
    ax.set_title("Figure 2: Women's Empowerment & Sustainability Indicators", fontweight='bold', pad=20)
    ax.set_xlim(0, 100)

    # Add value labels inside the bars
    for bar in bars:
        width = bar.get_width()
        ax.text(width - 5, bar.get_y() + bar.get_height()/2, f'{width}%',
                ha='right', va='center', color='white', fontweight='bold')

    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.savefig('Figure_2_Empowerment_Sustainability.png', dpi=300)
    plt.close()
    print("Figure 2 generated successfully.")

def generate_figure_3():
    """Generates the advanced aesthetic Flow Chart for the Conceptual Model."""
    dot = graphviz.Digraph(comment='Conceptual Model for Measuring Economic Returns', format='png')

    # Global Graph Settings for an advanced aesthetic
    dot.attr(rankdir='LR', splines='ortho', nodesep='0.6', ranksep='0.8')
    dot.attr('node', shape='box', style='filled,rounded', fontname='Helvetica', fontsize='12', margin='0.2,0.1')
    dot.attr('edge', color='#555555', penwidth='1.5', arrowsize='0.8')

    # Define Node Clusters and Styles
    # 1. Inputs/Context
    dot.node('Assets', 'Livelihood Assets &\nVulnerability Context\n(Human, Financial, Physical)', fillcolor='#e1f5fe', color='#0288d1', penwidth='2')

    # 2. Interventions
    dot.node('Interventions', 'Value Chain Interventions\n(SDI/PKSF Model)\nBackward & Forward Linkages', fillcolor='#e8f5e9', color='#388e3c', penwidth='2')

    # 3. Mediators (Gender & Nutrition)
    dot.node('Gender', 'Gendered Agency &\nEmpowerment\n(Decision-making, Loan Access)', fillcolor='#f3e5f5', color='#8e24aa', penwidth='2')
    dot.node('Nutrition', 'Nutrition-Sensitive\nPathways\n(Dietary Diversity, MDD-W)', fillcolor='#fff3e0', color='#f57c00', penwidth='2')

    # 4. Institutional
    dot.node('Institutions', 'Institutional & Market\nEnvironment\n(LSPs, Sustainability, Exit Strategies)', fillcolor='#fffde7', color='#fbc02d', penwidth='2')

    # 5. Core Outcome
    dot.node('Returns', 'Sustainable Economic Returns\n(Income, Profit Margins, ROI,\nEmployment, Resilience)', fillcolor='#eceff1', color='#455a64', penwidth='3', fontsize='14')

    # Define Relationships (Edges)
    dot.edge('Assets', 'Interventions', label=' Baseline constraints')
    dot.edge('Interventions', 'Returns', label=' Upgrading & Efficiency')

    dot.edge('Interventions', 'Gender')
    dot.edge('Gender', 'Returns', label=' Agency')
    dot.edge('Gender', 'Nutrition')
    dot.edge('Nutrition', 'Returns', label=' Well-being')

    dot.edge('Institutions', 'Returns', label=' Sustainability Framework')
    dot.edge('Institutions', 'Interventions', dir='both') # Two-way relationship

    # Render and save
    try:
        dot.render('Figure_3_Conceptual_Model', view=False, cleanup=True)
        print("Figure 3 generated successfully.")
    except Exception as e:
        print(f"Graphviz rendering failed. Ensure Graphviz is installed on your system. Error: {e}")

if __name__ == "__main__":
    print("Generating visual assets...")
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    print("All assets generated. Check your current directory for the PNG files.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import graphviz

# ==========================================
# AESTHETIC CONFIGURATION
# ==========================================
# Set a sleek, modern academic theme
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
custom_palette = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd"]
sns.set_palette(custom_palette)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['figure.dpi'] = 300 # High resolution for thesis printing

# ==========================================
# FIGURE 6.1: DEMOGRAPHICS (Gender vs Location)
# ==========================================
def plot_demographics():
    data = pd.DataFrame({
        'Location': ['Urban', 'Urban', 'Rural', 'Rural'],
        'Gender': ['Male', 'Female', 'Male', 'Female'],
        'Count': [180, 144, 123, 145] # Approximated from your percentages
    })

    fig, ax = plt.subplots(figsize=(8, 5))
    sns.barplot(data=data, x='Location', y='Count', hue='Gender', palette=['#2c3e50', '#e74c3c'], ax=ax)

    ax.set_title('Sample Distribution by Location and Gender (N=592)', pad=15)
    ax.set_ylabel('Number of Respondents')
    ax.set_xlabel('Geographic Location')
    sns.despine(left=True)
    plt.tight_layout()
    plt.savefig('Fig_6_1_Demographics.png')
    plt.close()

# ==========================================
# FIGURE 6.2: SERVICE PERFORMANCE
# ==========================================
def plot_service_performance():
    data = pd.DataFrame({
        'Service': ['MFS', 'MFS', 'MFS', 'Cards/QR', 'Cards/QR', 'Cards/QR', 'Internet Banking', 'Internet Banking', 'Internet Banking'],
        'Metric': ['Awareness (%)', 'Active Usage (%)', 'Satisfaction (out of 100)',
                   'Awareness (%)', 'Active Usage (%)', 'Satisfaction (out of 100)',
                   'Awareness (%)', 'Active Usage (%)', 'Satisfaction (out of 100)'],
        'Score': [97.8, 78.6, 83.8, 81.3, 42.3, 77.2, 64.7, 29.1, 70.2] # Converted 7-pt satisfaction to % for scale harmony
    })

    fig, ax = plt.subplots(figsize=(10, 6))
    sns.barplot(data=data, x='Service', y='Score', hue='Metric', palette='viridis', ax=ax)

    ax.set_title('Awareness, Usage, and Satisfaction by Service Channel', pad=15)
    ax.set_ylabel('Percentage / Scaled Score')
    ax.set_ylim(0, 110)
    ax.legend(title='Metrics', bbox_to_anchor=(1.05, 1), loc='upper left')
    sns.despine(left=True)
    plt.tight_layout()
    plt.savefig('Fig_6_2_Service_Performance.png')
    plt.close()

# ==========================================
# FIGURE 6.3: STRUCTURAL MODEL FLOWCHART (GRAPHVIZ)
# ==========================================
def generate_sem_flowchart():
    # Initialize a directed graph with aesthetic settings
    dot = graphviz.Digraph('SEM', format='png')
    dot.attr(rankdir='LR', size='10,6', dpi='300', fontname='Helvetica', bgcolor='transparent')

    # Node styles (Independent variables, Moderators, Dependent variables)
    dot.attr('node', shape='rect', style='filled, rounded', color='#34495e', fillcolor='#ecf0f1', fontname='Helvetica-Bold', fontcolor='#2c3e50', penwidth='2')

    # Independent Constructs
    dot.node('PE', 'Performance\nExpectancy')
    dot.node('EE', 'Effort\nExpectancy')
    dot.node('AQ', 'Agent\nQuality')
    dot.node('PR', 'Post-Fraud\nRisk Perception')
    dot.node('TR', 'Trust\nRestoration')

    # Dependent Constructs
    dot.attr('node', fillcolor='#d5f5e3', color='#27ae60') # Green tint for outcomes
    dot.node('BI', 'Behavioral\nIntention\n(R² = 0.68)')
    dot.node('UB', 'Usage\nBehavior\n(R² = 0.59)')

    # Moderator
    dot.attr('node', shape='ellipse', fillcolor='#fdebd0', color='#f39c12')
    dot.node('GP', 'Governance\nPerception')

    # Edge styles (Relationships)
    dot.attr('edge', fontname='Helvetica', fontsize='10', color='#7f8c8d', penwidth='1.5')

    # Add paths with Beta values and significance
    dot.edge('PE', 'BI', label=' β = 0.41***\n(Large Effect)', penwidth='3', color='#2c3e50')
    dot.edge('TR', 'BI', label=' β = 0.37***\n(Medium-Large)', penwidth='2.5', color='#2c3e50')
    dot.edge('EE', 'BI', label=' β = 0.29***')
    dot.edge('AQ', 'BI', label=' β = 0.26***')
    dot.edge('PR', 'BI', label=' β = -0.22***\n(Negative)', color='#c0392b', fontcolor='#c0392b') # Red for negative path

    dot.edge('BI', 'UB', label=' Strongly Predicted\n(Q² = 0.33)', penwidth='3')

    # Moderation path
    dot.edge('GP', 'TR', label=' Moderates (β = 0.14**)', style='dashed', color='#e67e22')

    # Render to file
    dot.render('Fig_6_3_Structural_Model', view=False, cleanup=True)

# ==========================================
# FIGURE 6.4: IPA MATRIX (Scatter Plot)
# ==========================================
def plot_ipa_matrix():
    # Data from Section 6.6
    challenges = [
        "Security, Fraud & Cyber Risks",
        "Digital Literacy & Divides",
        "Regulatory & Governance Issues",
        "Agent Network & Liquidity"
    ]
    importance = [6.48, 6.03, 5.98, 5.72]
    performance = [3.92, 4.48, 3.87, 4.61]

    fig, ax = plt.subplots(figsize=(9, 7))
    sns.scatterplot(x=performance, y=importance, s=300, color='#e74c3c', edgecolor='black', zorder=5, ax=ax)

    # Crosshairs (Mean values for quadrants)
    mean_imp = np.mean(importance)
    mean_perf = np.mean(performance)
    ax.axhline(mean_imp, color='gray', linestyle='--', zorder=1)
    ax.axvline(mean_perf, color='gray', linestyle='--', zorder=1)

    # Quadrant Labels
    ax.text(mean_perf - 0.2, mean_imp + 0.2, 'Concentrate Here\n(High Priority)', color='#c0392b', fontweight='bold', alpha=0.6)
    ax.text(mean_perf + 0.2, mean_imp + 0.2, 'Keep Up Good Work', color='#27ae60', fontweight='bold', alpha=0.6)
    ax.text(mean_perf - 0.2, mean_imp - 0.2, 'Low Priority', color='#7f8c8d', fontweight='bold', alpha=0.6)
    ax.text(mean_perf + 0.2, mean_imp - 0.2, 'Possible Overkill', color='#f39c12', fontweight='bold', alpha=0.6)

    # Annotate points
    for i, txt in enumerate(challenges):
        ax.annotate(txt, (performance[i], importance[i]), xytext=(10, 5), textcoords='offset points',
                    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", lw=1, alpha=0.8))

    ax.set_title('Importance-Performance Analysis (IPA) of Ecosystem Challenges', pad=15)
    ax.set_xlabel('Performance (Current Handling Capacity)')
    ax.set_ylabel('Importance (User Perception)')
    ax.set_xlim(3.5, 5.0)
    ax.set_ylim(5.5, 6.8)
    plt.tight_layout()
    plt.savefig('Fig_6_4_IPA_Matrix.png')
    plt.close()

# ==========================================
# FIGURE 6.5: INCLUSION PROJECTIONS
# ==========================================
def plot_projections():
    years = [2025, 2030, 2041]
    account_ownership = [58.4, 81.2, 92.7]
    active_usage = [42.3, 56.8, 74.3]
    gender_gap = [11.8, 6.4, 2.1]

    fig, ax1 = plt.subplots(figsize=(10, 6))

    # Primary Axis (Ownership & Usage)
    ax1.plot(years, account_ownership, marker='o', markersize=8, linewidth=3, label='Account Ownership (%)', color='#2980b9')
    ax1.plot(years, active_usage, marker='s', markersize=8, linewidth=3, label='Active Digital Usage (%)', color='#27ae60')
    ax1.set_xlabel('Year', fontweight='bold')
    ax1.set_ylabel('Percentage of Adult Population (%)', fontweight='bold')
    ax1.set_ylim(0, 100)
    ax1.set_xticks(years)

    # Secondary Axis (Gender Gap)
    ax2 = ax1.twinx()
    ax2.plot(years, gender_gap, marker='^', markersize=8, linewidth=3, linestyle=':', label='Gender Gap (Percentage Points)', color='#e74c3c')
    ax2.set_ylabel('Gender Gap (Percentage Points)', fontweight='bold', color='#e74c3c')
    ax2.set_ylim(0, 15)
    ax2.tick_params(axis='y', colors='#e74c3c')

    # Legends
    lines_1, labels_1 = ax1.get_legend_handles_labels()
    lines_2, labels_2 = ax2.get_legend_handles_labels()
    ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='center right')

    plt.title('Projected Digital Financial Inclusion vs. Gender Gap (2025–2041)', pad=15)
    sns.despine(right=False)
    plt.tight_layout()
    plt.savefig('Fig_6_5_Projections.png')
    plt.close()

# ==========================================
# EXECUTE ALL PLOTS
# ==========================================
if __name__ == "__main__":
    print("Generating demographic dashboard...")
    plot_demographics()

    print("Generating service performance chart...")
    plot_service_performance()

    print("Generating SEM flowchart via Graphviz...")
    generate_sem_flowchart()

    print("Generating IPA Matrix...")
    plot_ipa_matrix()

    print("Generating inclusion projections...")
    plot_projections()

    print("All thesis figures successfully generated in the current directory.")
```

```python
import graphviz

def create_structural_flowchart():
    # Initialize a directed graph with high-end aesthetic attributes
    dot = graphviz.Digraph(comment='Extended UTAUT2 Structural Model', format='png')
    dot.attr(rankdir='LR', size='10,6', dpi='300', fontname='Helvetica', bgcolor='white')

    # Define node styles
    node_attr = {
        'style': 'filled,rounded',
        'shape': 'box',
        'fontname': 'Helvetica-Bold',
        'fontsize': '12',
        'margin': '0.2,0.1',
        'penwidth': '1.5'
    }

    # Core Constructs (Blue)
    dot.attr('node', **node_attr, fillcolor='#E3F2FD', color='#1976D2', fontcolor='#0D47A1')
    dot.node('PE', 'Performance\nExpectancy\n(β=0.41***)')
    dot.node('EE', 'Effort\nExpectancy\n(β=0.29***)')

    # Post-Crisis & Institutional Constructs (Orange/Warning)
    dot.attr('node', **node_attr, fillcolor='#FFF3E0', color='#F57C00', fontcolor='#E65100')
    dot.node('TR', 'Trust\nRestoration\n(β=0.37***)')
    dot.node('GP', 'Governance\nPerception\n(Moderator β=0.14**)')
    dot.node('PFR', 'Post-Fraud\nRisk Perception\n(Females β=-0.31*)')

    # Rural Focus Construct (Green)
    dot.attr('node', **node_attr, fillcolor='#E8F5E9', color='#388E3C', fontcolor='#1B5E20')
    dot.node('AQ', 'Agent Quality\n(Rural β=0.38*)')

    # Dependent Variables (Dark Blue/Purple)
    dot.attr('node', **node_attr, fillcolor='#EDE7F6', color='#512DA8', fontcolor='#311B92')
    dot.node('BI', 'Behavioral\nIntention\n(R² = 0.68)')
    dot.node('UB', 'Usage\nBehavior\n(R² = 0.59)')

    # Edges with Beta labels and aesthetic formatting
    edge_attr = {'fontname': 'Helvetica', 'fontsize': '10', 'color': '#555555', 'penwidth': '1.5'}

    # Draw paths
    dot.edge('PE', 'BI', **edge_attr)
    dot.edge('EE', 'BI', **edge_attr)
    dot.edge('TR', 'BI', **edge_attr)
    dot.edge('AQ', 'BI', **edge_attr)
    dot.edge('PFR', 'BI', **edge_attr)

    # Draw Moderation (Governance -> Trust Restoration path)
    dot.edge('GP', 'TR', label=' Moderates ', style='dashed', **edge_attr)

    # BI to Usage
    dot.edge('BI', 'UB', label=' Strong Path ', **edge_attr)

    # Render the graph
    dot.render('Figure_7_1_Structural_Model', view=True)

if __name__ == "__main__":
    create_structural_flowchart()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def create_scenario_line_graph():
    # Set high-end aesthetic style
    plt.style.use('seaborn-v0_8-whitegrid')
    sns.set_context("talk", font_scale=0.9)

    # Data points (Years and Active Usage Percentages based on text)
    years = [2025, 2030, 2035, 2041]

    # Data extracted from Section 7.4
    optimistic = [42.3, 65.0, 80.0, 92.0]
    base = [42.3, 53.5, 58.0, 62.5]       # Midpoints of text ranges
    pessimistic = [42.3, 40.0, 39.0, 38.0] # Stagnation/Decline

    fig, ax = plt.subplots(figsize=(11, 7), dpi=300)

    # Plot lines with distinct, academic markers and colors
    ax.plot(years, optimistic, marker='o', markersize=8, linewidth=3,
            color='#2ca02c', label='Optimistic (High-Trust, Inclusive)')
    ax.plot(years, base, marker='s', markersize=8, linewidth=3,
            color='#1f77b4', label='Base (Moderate, Uneven Progress)')
    ax.plot(years, pessimistic, marker='^', markersize=8, linewidth=3,
            color='#d62728', label='Pessimistic (Trust Erosion, Fragmentation)')

    # Fill between to show divergence
    ax.fill_between(years, base, optimistic, color='#2ca02c', alpha=0.1)
    ax.fill_between(years, pessimistic, base, color='#d62728', alpha=0.05)

    # Annotations for key milestones
    ax.annotate('Smart Bangladesh 2041 Target', xy=(2041, 92), xytext=(2035, 88),
                arrowprops=dict(facecolor='#2ca02c', shrink=0.05, width=1.5, headwidth=6),
                fontsize=10, fontweight='bold', color='#1B5E20')

    # Formatting
    ax.set_title('Prospect Evaluation: Active Digital Usage Growth Scenarios (2025-2041)',
                 fontweight='bold', fontsize=16, pad=20)
    ax.set_xlabel('Year', fontweight='bold', fontsize=12)
    ax.set_ylabel('Active Digital Usage (%)', fontweight='bold', fontsize=12)

    ax.set_xticks(years)
    ax.set_ylim(30, 100)

    # Legend styling
    ax.legend(loc='center left', bbox_to_anchor=(1.02, 0.5), frameon=True,
              fancybox=True, shadow=True, title='Trajectories')

    # Clean spines
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    plt.tight_layout()
    plt.savefig('Figure_7_3_Scenarios_LineGraph.png', dpi=300, bbox_inches='tight')
    plt.show()

if __name__ == "__main__":
    create_scenario_line_graph()
```

```python
import graphviz

def create_roadmap_flowchart():
    # Initialize a directed graph with high-end academic aesthetics
    dot = graphviz.Digraph(comment='Implementation Roadmap', format='png')
    dot.attr(rankdir='TB', size='10,12', dpi='300', fontname='Helvetica', nodesep='0.6', ranksep='0.8')

    # Base node styles
    node_attr = {
        'shape': 'box', 'style': 'filled,rounded', 'fontname': 'Helvetica-Bold',
        'fontsize': '11', 'margin': '0.2,0.1', 'penwidth': '1.2'
    }

    # Create Subgraphs for each phase to maintain alignment and aesthetic grouping
    with dot.subgraph(name='cluster_short') as c_short:
        c_short.attr(style='rounded,filled', fillcolor='#F8F9FA', color='#CED4DA', penwidth='2')
        c_short.attr(label='Phase 1: Short-Term (0-2 Years) | Trust & Security', fontname='Helvetica-Bold', fontsize='14', fontcolor='#343A40')
        c_short.attr('node', **node_attr, fillcolor='#E3F2FD', color='#1976D2', fontcolor='#0D47A1')
        c_short.node('ST1', 'Mandatory Cybersecurity Audits\n& 72-Hour Redressal Protocol')
        c_short.node('ST2', 'National Digital Trust Score\nDashboard Launch')
        c_short.node('ST3', 'Mass Digital Literacy Drive\n(Target: 15M Adults)')
        c_short.node('ST4', 'Agent Certification Network\n(100% of 1.2M Agents)')

    with dot.subgraph(name='cluster_med') as c_med:
        c_med.attr(style='rounded,filled', fillcolor='#F8F9FA', color='#CED4DA', penwidth='2')
        c_med.attr(label='Phase 2: Medium-Term (2-5 Years) | Infrastructure & Interoperability', fontname='Helvetica-Bold', fontsize='14', fontcolor='#343A40')
        c_med.attr('node', **node_attr, fillcolor='#FFF3E0', color='#F57C00', fontcolor='#E65100')
        c_med.node('MT1', 'Full NPSB/IDTP Interoperability\n(90% Volume Target)')
        c_med.node('MT2', 'Digital Bank Operationalization\n(15 Approvals Active)')
        c_med.node('MT3', 'AI/Biometric Onboarding Scaled\n(50% of New Accounts)')
        c_med.node('MT4', 'Wholesale CBDC Sandbox\n& Retail Pilots')

    with dot.subgraph(name='cluster_long') as c_long:
        c_long.attr(style='rounded,filled', fillcolor='#F8F9FA', color='#CED4DA', penwidth='2')
        c_long.attr(label='Phase 3: Long-Term (5-10 Years) | Smart Ecosystem Integration', fontname='Helvetica-Bold', fontsize='14', fontcolor='#343A40')
        c_long.attr('node', **node_attr, fillcolor='#E8F5E9', color='#388E3C', fontcolor='#1B5E20')
        c_long.node('LT1', 'Near-Universal Formal Inclusion\n(92%+ Account Ownership)')
        c_long.node('LT2', 'Cashless Transition Achieved\n(Cash <20% of Retail)')
        c_long.node('LT3', 'Climate-Resilient Digital Finance\n(Green Micro-loans scaled)')
        c_long.node('LT4', 'South Asian Fintech Leadership\n& Cross-Border Linkages')

    # Add invisible edges to enforce top-to-bottom layout hierarchy
    edge_attr = {'color': '#6C757D', 'penwidth': '2', 'arrowsize': '0.8'}

    dot.edge('ST1', 'MT1', **edge_attr)
    dot.edge('ST2', 'MT2', **edge_attr)
    dot.edge('ST3', 'MT3', **edge_attr)
    dot.edge('ST4', 'MT4', **edge_attr)

    dot.edge('MT1', 'LT1', **edge_attr)
    dot.edge('MT2', 'LT2', **edge_attr)
    dot.edge('MT3', 'LT3', **edge_attr)
    dot.edge('MT4', 'LT4', **edge_attr)

    # Render
    dot.render('Figure_8_1_Implementation_Roadmap', view=True)

if __name__ == "__main__":
    create_roadmap_flowchart()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def create_sdg_barbell_chart():
    # Aesthetic setup
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, ax = plt.subplots(figsize=(10, 5), dpi=300)

    # Data extracted directly from Chapter 8 text metrics
    metrics = ['Active Digital Usage (%)', 'Gender Inclusion Gap (pp)', 'Rural-Urban Gap (pp)']
    current_2025 = [42.3, 11.8, 23.6]
    target_2030 = [65.0, 6.9, 9.9] # Using the <7 and <10 targets from the text

    y = np.arange(len(metrics))

    # Define colors representing current state vs optimized target
    color_current = '#d32f2f' # Red for current deficits
    color_target = '#2ca02c'  # Green for achieved targets
    line_color = '#b0bec5'

    # Draw lines connecting the points
    for i in range(len(metrics)):
        ax.plot([current_2025[i], target_2030[i]], [y[i], y[i]], color=line_color, zorder=1, linewidth=4)

    # Plot the points
    ax.scatter(current_2025, y, color=color_current, s=250, zorder=2, label='Current State (2025)')
    ax.scatter(target_2030, y, color=color_target, s=250, zorder=3, label='Optimistic Target (2030)')

    # Add data labels
    for i in range(len(metrics)):
        # Adjust label position based on which value is higher
        if current_2025[i] < target_2030[i]:
            ax.text(current_2025[i] - 1.5, y[i] + 0.15, f"{current_2025[i]}", color=color_current, fontweight='bold', ha='center')
            ax.text(target_2030[i] + 1.5, y[i] + 0.15, f"{target_2030[i]}", color=color_target, fontweight='bold', ha='center')
        else:
            ax.text(current_2025[i] + 1.5, y[i] + 0.15, f"{current_2025[i]}", color=color_current, fontweight='bold', ha='center')
            ax.text(target_2030[i] - 1.5, y[i] + 0.15, f"{target_2030[i]}", color=color_target, fontweight='bold', ha='center')

    # Formatting and aesthetics
    ax.set_yticks(y)
    ax.set_yticklabels(metrics, fontweight='bold', fontsize=11)
    ax.set_xlabel('Percentage (%) / Percentage Points (pp)', fontweight='bold', fontsize=11)
    ax.set_title('SDG 10 Alignment: Key Inclusion & Usage Targets (2025 vs 2030)', fontweight='bold', fontsize=14, pad=20)

    # Legend and Spines
    ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.25), ncol=2, frameon=False, fontsize=11)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.xaxis.grid(True, linestyle='--', alpha=0.5)
    ax.yaxis.grid(False)

    plt.tight_layout()
    plt.savefig('Figure_8_2_SDG_Barbell.png', dpi=300, bbox_inches='tight')
    plt.show()

if __name__ == "__main__":
    create_sdg_barbell_chart()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import graphviz
import os

# Set a modern, clean aesthetic for all matplotlib plots
sns.set_theme(style="whitegrid", context="talk", palette="deep")

def generate_conceptual_flowchart():
    """Generates an advanced, aesthetic flowchart for the Conceptual Model using Graphviz."""
    dot = graphviz.Digraph(comment='Conceptual Model', format='png')

    # Global aesthetics
    dot.attr(rankdir='LR', dpi='300', fontname='Helvetica, Arial, sans-serif')
    dot.attr('node', shape='rect', style='filled,rounded', fontname='Helvetica',
             fontsize='12', fontcolor='#333333', color='none', margin='0.2')
    dot.attr('edge', color='#888888', penwidth='1.5', arrowsize='0.8')

    # Define Node Clusters with soft pastel hex colors
    dot.node('A', 'Livelihood Assets\n(Human, Financial, Physical, Social)', fillcolor='#E3F2FD')
    dot.node('B', 'Value Chain Interventions\n(Backward & Forward Linkages)', fillcolor='#E8F5E9')
    dot.node('C', 'Gendered Agency\n& Empowerment', fillcolor='#FCE4EC')
    dot.node('D', 'Nutrition-Sensitive\nPathways (MDD-W)', fillcolor='#FFF3E0')
    dot.node('E', 'Institutional & Market\nEnvironment (SDI/PKSF)', fillcolor='#F3E5F5')
    dot.node('F', 'Sustainable Economic Returns\n(Income, Profit Margin, ROI, Resilience)',
             fillcolor='#E0F7FA', style='filled,rounded,bold', fontcolor='#006064')

    # Define flows/relationships
    dot.edge('E', 'B', label=' Facilitates')
    dot.edge('A', 'B', label=' Inputs')
    dot.edge('B', 'C', label=' Targets')
    dot.edge('B', 'D', label=' Generates Food/Income')
    dot.edge('C', 'D', label=' Decision Making')
    dot.edge('B', 'F', label=' Direct Returns')
    dot.edge('C', 'F', label=' Mediates')
    dot.edge('D', 'F', label=' Human Capital Gen.')

    # Render and save
    dot.render('Figure1_Conceptual_Flowchart', view=False)
    print("Generated: Figure1_Conceptual_Flowchart.png")

def generate_impact_bar_chart():
    """Generates a side-by-side bar chart for Income and MDD-W outcomes."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    fig.suptitle('Impact of VCD Interventions on Key Outcomes', fontsize=18, fontweight='bold', y=1.05)

    categories = ['Control Group', 'Project Group']

    # Left Plot: Income Baseline (Indexed at 100 for control to show 40% higher)
    income_data = [100, 140]
    sns.barplot(x=categories, y=income_data, ax=axes[0], palette=['#B0BEC5', '#2E7D32'])
    axes[0].set_title('Income Comparison (Indexed)', fontsize=14)
    axes[0].set_ylabel('Income Index (Control = 100)')
    axes[0].bar_label(axes[0].containers[0], fmt='%.0f', padding=3, fontsize=12)

    # Right Plot: MDD-W Percentages
    mddw_data = [72.06, 89.40]
    sns.barplot(x=categories, y=mddw_data, ax=axes[1], palette=['#B0BEC5', '#1565C0'])
    axes[1].set_title('Dietary Diversity (MDD-W Threshold)', fontsize=14)
    axes[1].set_ylabel('Percentage of Women (%)')
    axes[1].set_ylim(0, 100)
    axes[1].bar_label(axes[1].containers[0], fmt='%.1f%%', padding=3, fontsize=12)

    plt.tight_layout()
    plt.savefig('Figure2_Impact_BarChart.png', dpi=300, bbox_inches='tight')
    plt.close()
    print("Generated: Figure2_Impact_BarChart.png")

def generate_gender_radar_chart():
    """Generates a radar chart illustrating Kabeer's empowerment dimensions."""
    labels = np.array(['Asset Ownership\n(Resources)', 'Loan Access\n(Resources)',
                       'Decision-Making\n(Agency)', 'Market Participation\n(Agency)',
                       'Income Control\n(Achievements)'])
    num_vars = len(labels)

    # Conceptual data based on text (0-100 scale)
    baseline_data = [30, 40, 35, 20, 25]
    project_data = [65, 70, 78, 60, 70] # 78% decision making, 70% loan access per text

    # Compute angle for each axis
    angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
    baseline_data += baseline_data[:1]
    project_data += project_data[:1]
    angles += angles[:1]

    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
    ax.set_theta_offset(np.pi / 2)
    ax.set_theta_direction(-1)

    # Draw axes and labels
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels, fontsize=11, fontweight='bold')

    # Plot baseline
    ax.plot(angles, baseline_data, color='#9E9E9E', linewidth=2, linestyle='solid', label='Pre-Intervention (Control Baseline)')
    ax.fill(angles, baseline_data, color='#9E9E9E', alpha=0.25)

    # Plot project
    ax.plot(angles, project_data, color='#D81B60', linewidth=2, linestyle='solid', label='Post-Intervention (Project Group)')
    ax.fill(angles, project_data, color='#D81B60', alpha=0.25)

    ax.set_title('Gendered Economic Empowerment Dynamics', size=16, fontweight='bold', pad=30)
    ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

    plt.tight_layout()
    plt.savefig('Figure3_Gender_RadarChart.png', dpi=300, bbox_inches='tight')
    plt.close()
    print("Generated: Figure3_Gender_RadarChart.png")

if __name__ == "__main__":
    generate_conceptual_flowchart()
    generate_impact_bar_chart()
    generate_gender_radar_chart()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def create_flowchart():
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.axis('off')

    # Define colors
    primary_blue = '#1A5276'
    accent_green = '#1D8348'
    bg_color = '#F4F6F7'

    # Draw Background
    fig.patch.set_facecolor(bg_color)

    # Helper function for boxes
    def draw_box(x, y, text, color, width=2.5, height=1.2):
        rect = patches.FancyBboxPatch((x, y), width, height, boxstyle="round,pad=0.1",
                                      linewidth=2, edgecolor=color, facecolor='white', zorder=3)
        ax.add_patch(rect)
        ax.text(x + width/2, y + height/2, text, ha='center', va='center',
                fontsize=10, fontweight='bold', color=color, wrap=True)

    # 1. Foundation Layer
    draw_box(0.5, 7.5, "Capacity Building\n(GAP & LSP Training)", primary_blue)
    draw_box(3.7, 7.5, "Input Systems\n(Fodder & Vaccine Hubs)", primary_blue)
    draw_box(7.0, 7.5, "Technology Adoption\n(Milking & Processing)", primary_blue)

    # 2. Integration Layer (The Core Strategy)
    draw_box(3.7, 4.5, "Integrated Value Chain\nIntervention Strategy", accent_green, width=2.5, height=1.5)

    # 3. Outcome Layer
    draw_box(0.5, 1.5, "Market Linkage &\nBranding", primary_blue)
    draw_box(3.7, 1.5, "Eco-Friendly\nSustainability", primary_blue)
    draw_box(7.0, 1.5, "Women's Economic\nEmpowerment", primary_blue)

    # Draw Arrows
    arrow_style = dict(arrowstyle='->', color='gray', lw=1.5, mutation_scale=20)
    ax.annotate('', xy=(1.75, 7.5), xytext=(4.5, 6.0), arrowprops=arrow_style) # From Top Left
    ax.annotate('', xy=(5.0, 7.5), xytext=(5.0, 6.0), arrowprops=arrow_style)  # From Top Mid
    ax.annotate('', xy=(8.25, 7.5), xytext=(5.5, 6.0), arrowprops=arrow_style) # From Top Right

    ax.annotate('', xy=(1.75, 2.7), xytext=(4.5, 4.5), arrowprops=arrow_style) # To Bottom Left
    ax.annotate('', xy=(5.0, 2.7), xytext=(5.0, 4.5), arrowprops=arrow_style)  # To Bottom Mid
    ax.annotate('', xy=(8.25, 2.7), xytext=(5.5, 4.5), arrowprops=arrow_style) # To Bottom Right

    plt.title("Figure 3.2: SDI Integrated Intervention Strategy Framework", fontsize=16, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.show()

create_flowchart()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from graphviz import Digraph

# Set advanced aesthetic styling for all plots
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
custom_palette = ["#2A9D8F", "#E9C46A", "#F4A261", "#E76F51", "#264653"]
sns.set_palette(custom_palette)

def save_plot(filename):
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches='tight', transparent=False, facecolor='white')
    plt.close()

# ---------------------------------------------------------
# Figure 1: Gender Distribution (Grouped Bar Chart)
# ---------------------------------------------------------
def plot_gender_distribution():
    data = {
        'Group': ['Project Respondent', 'Project Respondent', 'Control Respondent', 'Control Respondent'],
        'Gender': ['Male', 'Female', 'Male', 'Female'],
        'Percentage': [8, 92, 0, 100]
    }
    df = pd.DataFrame(data)

    plt.figure(figsize=(8, 6))
    ax = sns.barplot(x='Group', y='Percentage', hue='Gender', data=df, palette=["#264653", "#E76F51"])
    plt.title('Figure 1: Gender Distribution of Participants', fontsize=16, weight='bold', pad=20)
    plt.ylabel('Percentage (%)', fontsize=12, weight='bold')
    plt.xlabel('', fontsize=12)
    plt.ylim(0, 110)

    # Add data labels
    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f"{int(p.get_height())}%",
                        (p.get_x() + p.get_width() / 2., p.get_height()),
                        ha='center', va='bottom', fontsize=12, weight='bold', xytext=(0, 5),
                        textcoords='offset points')

    save_plot('Fig1_Gender_Distribution.png')

# ---------------------------------------------------------
# Figure 2: Age Profile (Grouped Bar Chart)
# ---------------------------------------------------------
def plot_age_profile():
    data = {
        'Age Range': ['18–35', '36–49', '50–60', '61+'] * 2,
        'Group': ['Project'] * 4 + ['Control'] * 4,
        'Percentage': [61, 36, 4, 0, 56, 41, 3, 0]
    }
    df = pd.DataFrame(data)

    plt.figure(figsize=(10, 6))
    ax = sns.barplot(x='Age Range', y='Percentage', hue='Group', data=df, palette=["#2A9D8F", "#F4A261"])
    plt.title('Figure 2: Age Profile of Respondents', fontsize=16, weight='bold', pad=20)
    plt.ylabel('Percentage (%)', fontsize=12, weight='bold')
    plt.xlabel('Age Range (Years)', fontsize=12, weight='bold')
    plt.ylim(0, 70)

    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f"{int(p.get_height())}%",
                        (p.get_x() + p.get_width() / 2., p.get_height()),
                        ha='center', va='bottom', fontsize=11, xytext=(0, 5),
                        textcoords='offset points')

    save_plot('Fig2_Age_Profile.png')

# ---------------------------------------------------------
# Figure 3: Household Monthly Income (Line/Area Plot or Grouped Bar)
# ---------------------------------------------------------
def plot_income_distribution():
    data = {
        'Income Bracket': ['8k–15k', '15k–25k', '25k–35k', '35k–45k', '45k+'] * 2,
        'Group': ['Project'] * 5 + ['Control'] * 5,
        'Percentage': [6.7, 28.7, 52.2, 9.1, 3.2, 33.8, 47.1, 10.3, 5.9, 2.9]
    }
    df = pd.DataFrame(data)

    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='Income Bracket', y='Percentage', hue='Group', data=df, palette=["#264653", "#E9C46A"])
    plt.title('Figure 3: Household Monthly Income Distribution', fontsize=16, weight='bold', pad=20)
    plt.ylabel('Percentage of Households (%)', fontsize=12, weight='bold')
    plt.xlabel('Income Bracket (Tk.)', fontsize=12, weight='bold')

    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f"{p.get_height()}%",
                        (p.get_x() + p.get_width() / 2., p.get_height()),
                        ha='center', va='bottom', fontsize=10, xytext=(0, 5),
                        textcoords='offset points')

    save_plot('Fig3_Income_Distribution.png')

# ---------------------------------------------------------
# Figure 4: Future Cattle Rearing Plans (Aesthetic Donut Chart)
# ---------------------------------------------------------
def plot_future_plans():
    labels = ['Expanding business', 'No change', 'Shifting/Shut down']
    sizes = [74, 25, 1]
    colors = ['#2A9D8F', '#E9C46A', '#E76F51']
    explode = (0.05, 0, 0)

    fig, ax = plt.subplots(figsize=(8, 8))
    wedges, texts, autotexts = ax.pie(sizes, explode=explode, labels=labels, colors=colors,
                                      autopct='%1.0f%%', startangle=140, pctdistance=0.85,
                                      textprops=dict(color="black", fontsize=12, weight="bold"))

    # Draw circle for Donut effect
    centre_circle = plt.Circle((0, 0), 0.70, fc='white')
    fig.gca().add_artist(centre_circle)

    plt.title('Figure 4: Future Cattle Rearing Plans', fontsize=16, weight='bold', pad=20)
    save_plot('Fig4_Future_Plans.png')

# ---------------------------------------------------------
# Figure 5: Herd Size (Clustered Bar Chart)
# ---------------------------------------------------------
def plot_herd_size():
    data = {
        'Herd Size': ['1 to 2', '3 to 5', '5 and above'] * 2,
        'Group': ['Project'] * 3 + ['Control'] * 3,
        'Percentage': [18, 54, 28, 32, 49, 19]
    }
    df = pd.DataFrame(data)

    plt.figure(figsize=(9, 6))
    ax = sns.barplot(x='Herd Size', y='Percentage', hue='Group', data=df, palette=["#E76F51", "#264653"])
    plt.title('Figure 5: Herd Size Distribution', fontsize=16, weight='bold', pad=20)
    plt.ylabel('Percentage (%)', fontsize=12, weight='bold')
    plt.xlabel('Number of Cattle', fontsize=12, weight='bold')

    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f"{int(p.get_height())}%",
                        (p.get_x() + p.get_width() / 2., p.get_height()),
                        ha='center', va='bottom', fontsize=11, xytext=(0, 5),
                        textcoords='offset points')

    save_plot('Fig5_Herd_Size.png')

# ---------------------------------------------------------
# Figure 6: Flowchart of Interventions (Graphviz)
# ---------------------------------------------------------
def create_flowchart():
    # Requires graphviz to be installed on your system and Python environment
    dot = Digraph(comment='RMTP Intervention Flow', format='png')
    dot.attr(rankdir='TB', size='10,8', dpi='300', fontname='Helvetica', splines='ortho')

    # Define aesthetic node styles
    dot.attr('node', shape='rect', style='filled,rounded', color='#264653', fontcolor='white', fontname='Helvetica-Bold', fontsize='12', margin='0.3')

    dot.node('A', 'Target Demographic\n(92% Female, Prime Working Age)')
    dot.node('B', 'Microfinance Access\n(Agrosor & Jagoron Loans)')
    dot.node('C', 'Livestock Enterprise Setup\n(Cow-Calf, Beef Fattening, Mixed)')

    dot.attr('node', color='#2A9D8F', fontcolor='white')
    dot.node('D', 'Scale-up Operations\n(Herd Size Expansion to 3-5+)')
    dot.node('E', 'Economic Impact\n(+40% Monthly Income vs Control)')

    dot.attr('node', color='#E76F51', fontcolor='white')
    dot.node('F', 'Future Outlook\n(74% Plan to Expand Business)')

    # Edges
    dot.attr('edge', color='#555555', penwidth='2')
    dot.edge('A', 'B')
    dot.edge('B', 'C')
    dot.edge('C', 'D')
    dot.edge('D', 'E')
    dot.edge('E', 'F')

    # Render flowchart
    dot.render('Fig6_Flowchart', view=False, cleanup=True)

# ---------------------------------------------------------
# Execution
# ---------------------------------------------------------
if __name__ == "__main__":
    plot_gender_distribution()
    plot_age_profile()
    plot_income_distribution()
    plot_future_plans()
    plot_herd_size()

    try:
        create_flowchart()
        print("All visual artifacts, including the flowchart, generated successfully.")
    except Exception as e:
        print(f"Figures 1-5 generated. Flowchart failed (ensure Graphviz is installed on your OS): {e}")
```

```python
from graphviz import Digraph

def create_value_chain_flowchart():
    # Initialize Digraph with advanced aesthetics
    dot = Digraph(comment='Value Chain and Sustainability', format='png')
    dot.attr(rankdir='LR', size='12,8', dpi='300', fontname='Helvetica', outputorder='edgesfirst')

    # Define node styles
    dot.attr('node', shape='rect', style='filled,rounded', fontname='Helvetica-Bold', fontcolor='white', penwidth='0')
    dot.attr('edge', color='#555555', penwidth='2', arrowsize='0.8')

    # 1. Input Supply Nodes (Blue)
    dot.attr('node', fillcolor='#1f77b4')
    dot.node('A1', 'Input Supply System\n(100% Linkage)')
    dot.node('A2', 'Vaccine Hubs & LSPs\n(Decentralised)')
    dot.node('A3', 'Feed & Nutrition\n(Napier, Mixed Grains)')

    # 2. Production & Eco-Practices Nodes (Green)
    dot.attr('node', fillcolor='#2ca02c')
    dot.node('B1', 'Livestock Producers\n(GAP Adopted)')
    dot.node('B2', 'Eco-Friendly Practices\n(91-95% Adoption)')
    dot.node('B3', 'Vermicompost &\nWaste Management')

    # 3. Market Linkage Nodes (Orange)
    dot.attr('node', fillcolor='#ff7f0e')
    dot.node('C1', 'Market Linkages\n(Collective Marketing)')
    dot.node('C2', 'Milk Collection Points\n(17% Established)')
    dot.node('C3', 'Higher Sales Volume\n(80% of Participants)')

    # 4. Sustainability Nodes (Purple)
    dot.attr('node', fillcolor='#9467bd')
    dot.node('D1', 'Institutional Sustainability')
    dot.node('D2', 'Local Service Ecosystem\n(LSPs, Producer Groups)')

    # Create subgraphs to align nodes neatly
    with dot.subgraph() as s:
        s.attr(rank='same')
        s.edge('A2', 'A1', style='invis')
        s.edge('A3', 'A1', style='invis')

    # Define the flows (Edges)
    dot.edge('A2', 'B1', label=' Health Services')
    dot.edge('A3', 'B1', label=' Nutrition inputs')
    dot.edge('A1', 'B1', label=' 100% Access')

    dot.edge('B1', 'B2', label=' Tech Adoption')
    dot.edge('B2', 'B3')

    dot.edge('B1', 'C1', label=' Goods/Products')
    dot.edge('C1', 'C2')
    dot.edge('C1', 'C3')

    dot.edge('C3', 'D1', label=' Profitability')
    dot.edge('B1', 'D2', label=' Capacity Building')
    dot.edge('D2', 'D1', label=' Handover Strategy')

    # Render the flowchart to a file
    dot.render('Figure_3_Value_Chain_Flowchart', view=True)

# Generate the flowchart
create_value_chain_flowchart()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set aesthetic style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 7))
fig.suptitle('Economic Impact of Drug Policy: Interdiction vs. Education', fontsize=18, fontweight='bold', y=1.05)

# --- Graph 1: Supply-Side Interdiction (Friedman's Critique) ---
ax1 = axes[0]
q = np.linspace(1, 10, 100)

# Inelastic Demand (Steep)
demand_inelastic = 12 - 0.8 * q
# Original Supply
supply_1 = 2 + 0.5 * q
# New Supply (Shifted Left due to Interdiction)
supply_2 = 5 + 0.5 * q

# Plot curves
ax1.plot(q, demand_inelastic, color='#E63946', linewidth=3, label='Demand (Inelastic)')
ax1.plot(q, supply_1, color='#457B9D', linewidth=3, label='Supply 1 (Baseline)')
ax1.plot(q, supply_2, color='#1D3557', linewidth=3, linestyle='--', label='Supply 2 (Post-Interdiction)')

# Intersections (P, Q)
eq1_q, eq1_p = 7.69, 5.85
eq2_q, eq2_p = 5.38, 7.69

# Revenue Boxes
ax1.add_patch(plt.Rectangle((0, 0), eq1_q, eq1_p, color='#A8DADC', alpha=0.3, label='Original Revenue'))
ax1.add_patch(plt.Rectangle((0, 0), eq2_q, eq2_p, fill=False, edgecolor='#1D3557', hatch='//', linewidth=2, label='New Revenue (Increased)'))

# Annotations & Formatting
ax1.plot(eq1_q, eq1_p, 'ko', markersize=8)
ax1.plot(eq2_q, eq2_p, 'ko', markersize=8)
ax1.annotate('Huge Price Spike', xy=(eq2_q, eq2_p), xytext=(eq2_q+1, eq2_p+1),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8), fontsize=12)

ax1.set_title('Figure 1: The Interdiction Paradox', fontsize=14)
ax1.set_xlabel('Quantity (Q)', fontsize=12)
ax1.set_ylabel('Price (P)', fontsize=12)
ax1.set_xlim(0, 10)
ax1.set_ylim(0, 12)
ax1.legend(loc='upper right')

# --- Graph 2: Demand-Side Education & Rehab (Harm Reduction) ---
ax2 = axes[1]

# Original Demand (Inelastic)
demand_1 = 12 - 0.8 * q
# New Demand (Shifted Left due to Rehab/Education)
demand_2 = 8 - 0.8 * q
# Static Supply
supply_static = 2 + 0.5 * q

# Plot curves
ax2.plot(q, demand_1, color='#E63946', linewidth=3, label='Demand 1 (Baseline)')
ax2.plot(q, demand_2, color='#F4A261', linewidth=3, linestyle='--', label='Demand 2 (Post-Education)')
ax2.plot(q, supply_static, color='#457B9D', linewidth=3, label='Supply')

# Intersections (P, Q)
eq3_q, eq3_p = 7.69, 5.85
eq4_q, eq4_p = 4.61, 4.31

# Revenue Boxes
ax2.add_patch(plt.Rectangle((0, 0), eq3_q, eq3_p, color='#A8DADC', alpha=0.3, label='Original Revenue'))
ax2.add_patch(plt.Rectangle((0, 0), eq4_q, eq4_p, fill=False, edgecolor='#F4A261', hatch='\\\\', linewidth=2, label='New Revenue (Collapsed)'))

# Annotations & Formatting
ax2.plot(eq3_q, eq3_p, 'ko', markersize=8)
ax2.plot(eq4_q, eq4_p, 'ko', markersize=8)
ax2.annotate('Price & Quantity Collapse', xy=(eq4_q, eq4_p), xytext=(eq4_q+1, eq4_p+2),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8), fontsize=12)

ax2.set_title('Figure 2: Demand Starvation', fontsize=14)
ax2.set_xlabel('Quantity (Q)', fontsize=12)
ax2.set_ylabel('Price (P)', fontsize=12)
ax2.set_xlim(0, 10)
ax2.set_ylim(0, 12)
ax2.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set a modern, clean aesthetic style
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)

# Create a figure with 3 subplots in a row
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle('Socio-Economic & Demographic Analysis of Drug Policy in Bangladesh',
             fontsize=18, fontweight='bold', y=1.05)

# ---------------------------------------------------------
# Figure 3: Budget Reallocation (Current vs. Proposed)
# ---------------------------------------------------------
ax1 = axes[0]
categories = ['Law Enforcement\n& Border (BGB/RAB)', 'Judicial System\n& Prisons', 'Rehabilitation\n& Treatment', 'Education &\nPrevention', 'Socio-Economic\nIntegration']

# Synthetic data representing percentage of drug-control budget
current_policy = [65, 20, 10, 5, 0]
proposed_policy = [25, 10, 35, 15, 15]

x = np.arange(len(categories))
width = 0.35

rects1 = ax1.bar(x - width/2, current_policy, width, label='Current (Punitive)', color='#E63946', alpha=0.9)
rects2 = ax1.bar(x + width/2, proposed_policy, width, label='Proposed (Public Health)', color='#2A9D8F', alpha=0.9)

ax1.set_ylabel('Percentage of Fiscal Budget (%)', fontweight='bold')
ax1.set_title('Fig 3: Recommended Fiscal Reallocation', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(categories, rotation=45, ha="right")
ax1.legend()

# ---------------------------------------------------------
# Figure 4: The Crime-Price Nexus (Dual Axis Line Graph)
# ---------------------------------------------------------
ax2 = axes[1]
years = np.arange(2015, 2024)

# Synthetic data showing rising price due to interdiction and rising crime
yaba_price_bdt = [150, 200, 250, 350, 450, 500, 650, 700, 850]
property_crime_index = [100, 105, 115, 130, 145, 150, 175, 185, 210]

color1 = '#1D3557'
ax2.set_xlabel('Year', fontweight='bold')
ax2.set_ylabel('Avg. Yaba Street Price (BDT)', color=color1, fontweight='bold')
line1 = ax2.plot(years, yaba_price_bdt, color=color1, linewidth=3, marker='o', label='Yaba Price')
ax2.tick_params(axis='y', labelcolor=color1)

# Create a second y-axis for the crime index
ax2_twin = ax2.twinx()
color2 = '#E63946'
ax2_twin.set_ylabel('Property Crime Index (Base=100)', color=color2, fontweight='bold')
line2 = ax2_twin.plot(years, property_crime_index, color=color2, linewidth=3, linestyle='--', marker='s', label='Property Crime')
ax2_twin.tick_params(axis='y', labelcolor=color2)

# Combine legends
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax2.legend(lines, labels, loc='upper left')

ax2.set_title('Fig 4: Price-Crime Correlation (The Interdiction Effect)', fontsize=14, fontweight='bold')
ax2.grid(False) # Turn off grid to avoid clutter with dual axes

# ---------------------------------------------------------
# Figure 5: Demographic Vulnerability (Histogram/Bar)
# ---------------------------------------------------------
ax3 = axes[2]
age_groups = ['10-14', '15-19', '20-24', '25-29', '30-34', '35-39', '40+']
# Synthetic prevalence data representing the crisis in human capital
prevalence_percentage = [2, 18, 35, 25, 10, 6, 4]

bars = ax3.bar(age_groups, prevalence_percentage, color='#457B9D', edgecolor='black', linewidth=1.2)

# Highlight the most vulnerable groups (15-29) representing lost human capital
for i in [1, 2, 3]:
    bars[i].set_color('#E63946')
    bars[i].set_edgecolor('black')

ax3.set_ylabel('Percentage of Total User Base (%)', fontweight='bold')
ax3.set_xlabel('Age Group', fontweight='bold')
ax3.set_title('Fig 5: Human Capital Drain (Age Distribution)', fontsize=14, fontweight='bold')

# Add value labels on top of the bars
for bar in bars:
    yval = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2, yval + 0.5, f'{yval}%', ha='center', va='bottom', fontsize=10)

# ---------------------------------------------------------
# Final Formatting
# ---------------------------------------------------------
sns.despine(left=True, bottom=True) # Clean up borders
plt.tight_layout()
plt.subplots_adjust(top=0.88) # Adjust top to fit the main title
plt.show()
```

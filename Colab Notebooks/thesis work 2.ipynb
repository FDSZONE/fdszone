```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set global style for academic reporting
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12, 'figure.dpi': 300})

def generate_figure_7_1():
    """Figure 7.1: Urban vs. Rural Meaningful Internet Usage"""
    categories = ['Urban Centers', 'Rural Chuadanga']
    usage_percentages = [75, 33]  # Illustrating the 42% deficit mentioned in the text

    fig, ax = plt.subplots(figsize=(7, 5))
    bars = ax.bar(categories, usage_percentages, color=['#4C72B0', '#DD8452'], width=0.5)

    ax.set_ylabel('Meaningful Internet Usage (%)')
    ax.set_title('Figure 7.1: The Rural-Urban Digital Divide in Meaningful Usage')
    ax.set_ylim(0, 100)

    # Add percentage labels
    for bar in bars:
        yval = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, yval + 2, f'{yval}%', ha='center', va='bottom', fontweight='bold')

    plt.tight_layout()
    plt.savefig('Figure_7_1_Urban_Rural_Divide.png')
    plt.show()

def generate_figure_7_3():
    """Figure 7.3: Gender Gap in Mobile Internet Usage"""
    labels = ['Mobile Ownership', 'Active Internet Usage', 'MFS Account Ownership']
    male_data = [85, 68, 70]
    female_data = [55, 28, 35] # Illustrating the ~40% disparity

    x = np.arange(len(labels))
    width = 0.35

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.bar(x - width/2, male_data, width, label='Male', color='#55A868')
    ax.bar(x + width/2, female_data, width, label='Female', color='#C44E52')

    ax.set_ylabel('Adoption Rate (%)')
    ax.set_title('Figure 7.3: Gender Disparities in Digital and Financial Inclusion')
    ax.set_xticks(x)
    ax.set_xticklabels(labels)
    ax.set_ylim(0, 100)
    ax.legend()

    plt.tight_layout()
    plt.savefig('Figure_7_3_Gender_Divide.png')
    plt.show()

def generate_figure_8_2():
    """Figure 8.2: 60-Month Implementation Roadmap (Gantt Chart)"""
    # Phases and their start/duration in months
    phases = [
        "Phase 1: Infrastructural Stabilization",
        "Phase 2: Literacy & Access",
        "Phase 3: E-Agriculture Integration",
        "Phase 4: Ecosystem Autonomy"
    ]
    start_months = [1, 12, 24, 48]
    durations = [11, 12, 24, 12]

    fig, ax = plt.subplots(figsize=(10, 4))

    # Create horizontal bars for the Gantt chart
    colors = ['#4C72B0', '#55A868', '#C44E52', '#8172B3']
    for i, phase in enumerate(phases):
        ax.barh(phase, durations[i], left=start_months[i], color=colors[i], edgecolor='black')

    ax.set_xlabel('Implementation Timeline (Months)')
    ax.set_title('Figure 8.2: 60-Month Implementation Roadmap for the "Chuadanga Model"')
    ax.set_xlim(0, 60)
    ax.set_xticks(np.arange(0, 61, 12)) # Tick marks every 12 months (1 year)

    # Invert y-axis to have Phase 1 at the top
    ax.invert_yaxis()

    plt.tight_layout()
    plt.savefig('Figure_8_2_Implementation_Roadmap.png')
    plt.show()

# Execute functions to generate and save charts
generate_figure_7_1()
generate_figure_7_3()
generate_figure_8_2()
```

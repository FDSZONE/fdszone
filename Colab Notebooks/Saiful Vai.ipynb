```python
from graphviz import Digraph

def create_value_chain_flowchart():
    # Initialize Digraph with advanced aesthetics
    dot = Digraph(comment='Value Chain and Sustainability', format='png')
    dot.attr(rankdir='LR', size='12,8', dpi='300', fontname='Helvetica', outputorder='edgesfirst')

    # Define node styles
    dot.attr('node', shape='rect', style='filled,rounded', fontname='Helvetica-Bold', fontcolor='white', penwidth='0')
    dot.attr('edge', color='#555555', penwidth='2', arrowsize='0.8')

    # 1. Input Supply Nodes (Blue)
    dot.attr('node', fillcolor='#1f77b4')
    dot.node('A1', 'Input Supply System\n(100% Linkage)')
    dot.node('A2', 'Vaccine Hubs & LSPs\n(Decentralised)')
    dot.node('A3', 'Feed & Nutrition\n(Napier, Mixed Grains)')

    # 2. Production & Eco-Practices Nodes (Green)
    dot.attr('node', fillcolor='#2ca02c')
    dot.node('B1', 'Livestock Producers\n(GAP Adopted)')
    dot.node('B2', 'Eco-Friendly Practices\n(91-95% Adoption)')
    dot.node('B3', 'Vermicompost &\nWaste Management')

    # 3. Market Linkage Nodes (Orange)
    dot.attr('node', fillcolor='#ff7f0e')
    dot.node('C1', 'Market Linkages\n(Collective Marketing)')
    dot.node('C2', 'Milk Collection Points\n(17% Established)')
    dot.node('C3', 'Higher Sales Volume\n(80% of Participants)')

    # 4. Sustainability Nodes (Purple)
    dot.attr('node', fillcolor='#9467bd')
    dot.node('D1', 'Institutional Sustainability')
    dot.node('D2', 'Local Service Ecosystem\n(LSPs, Producer Groups)')

    # Create subgraphs to align nodes neatly
    with dot.subgraph() as s:
        s.attr(rank='same')
        s.edge('A2', 'A1', style='invis')
        s.edge('A3', 'A1', style='invis')

    # Define the flows (Edges)
    dot.edge('A2', 'B1', label=' Health Services')
    dot.edge('A3', 'B1', label=' Nutrition inputs')
    dot.edge('A1', 'B1', label=' 100% Access')

    dot.edge('B1', 'B2', label=' Tech Adoption')
    dot.edge('B2', 'B3')

    dot.edge('B1', 'C1', label=' Goods/Products')
    dot.edge('C1', 'C2')
    dot.edge('C1', 'C3')

    dot.edge('C3', 'D1', label=' Profitability')
    dot.edge('B1', 'D2', label=' Capacity Building')
    dot.edge('D2', 'D1', label=' Handover Strategy')

    # Render the flowchart to a file
    dot.render('Figure_3_Value_Chain_Flowchart', view=True)

# Generate the flowchart
create_value_chain_flowchart()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Set advanced aesthetic style
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
plt.rcParams['font.family'] = 'sans-serif'

def create_input_supply_chart():
    # Data for Section 9.1
    categories = ['Feed Suppliers', 'Medicine/Vaccine', 'Breeding/AI & LSPs']
    project_data = [100, 100, 100]
    control_data = [29, 29, 25]

    x = np.arange(len(categories))
    width = 0.35

    fig, ax = plt.subplots(figsize=(10, 6))
    rects1 = ax.bar(x - width/2, project_data, width, label='Project Participants', color='#2ca02c', edgecolor='black')
    rects2 = ax.bar(x + width/2, control_data, width, label='Control Group', color='#7f7f7f', edgecolor='black')

    # Formatting
    ax.set_ylabel('Linkage Achievement (%)', fontweight='bold')
    ax.set_title('Input Supply Linkage Comparison', fontsize=16, fontweight='bold', pad=15)
    ax.set_xticks(x)
    ax.set_xticklabels(categories, fontweight='bold')
    ax.legend(loc='lower right')
    ax.set_ylim(0, 115)

    # Auto-label bars
    ax.bar_label(rects1, padding=3, fmt='%d%%')
    ax.bar_label(rects2, padding=3, fmt='%d%%')

    plt.tight_layout()
    plt.savefig('Figure_1_Input_Supply.png', dpi=300)
    plt.show()

def create_eco_friendly_chart():
    # Data for Section 9.3
    categories = ['Eco-Friendly Rearing', 'Climate-Resilient\nPractices', 'Eco-Fertiliser\n(Vermicompost)']
    project_data = [91, 95, 92]
    control_data = [25, 37, 43]

    x = np.arange(len(categories))
    width = 0.35

    fig, ax = plt.subplots(figsize=(10, 6))
    rects1 = ax.bar(x - width/2, project_data, width, label='Project Participants', color='#1f77b4', edgecolor='black')
    rects2 = ax.bar(x + width/2, control_data, width, label='Control Group', color='#ff7f0e', edgecolor='black')

    # Formatting
    ax.set_ylabel('Adoption Rate (%)', fontweight='bold')
    ax.set_title('Adoption of Eco-Friendly and Climate-Resilient Practices', fontsize=16, fontweight='bold', pad=15)
    ax.set_xticks(x)
    ax.set_xticklabels(categories, fontweight='bold')
    ax.legend(loc='upper right')
    ax.set_ylim(0, 115)

    # Auto-label bars
    ax.bar_label(rects1, padding=3, fmt='%d%%')
    ax.bar_label(rects2, padding=3, fmt='%d%%')

    plt.tight_layout()
    plt.savefig('Figure_2_Eco_Practices.png', dpi=300)
    plt.show()

# Generate the charts
create_input_supply_chart()
create_eco_friendly_chart()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

def create_impact_dashboard():
    # Apply modern aesthetic style
    sns.set_theme(style="whitegrid", rc={"axes.spines.top": False, "axes.spines.right": False})

    # Data derived from the text
    data = {
        'Metric': ['Monthly Income (BDT)', 'GAP Adoption (%)', 'Vaccine Use (%)', 'MDD-W Target Met (%)'],
        'Project': [32161, 61, 100, 89.4],
        'Control': [22919, 21, 30, 72.1]
    }
    df = pd.DataFrame(data)

    # Set up a 2x2 grid for the dashboard
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    fig.suptitle('Project Impact Comparison: Baseline vs Control', fontsize=18, fontweight='bold', y=0.95)

    colors = ['#2A9D8F', '#E76F51'] # Modern Teal and Terracotta palette

    # Helper function to plot individual bar charts
    def plot_bar(ax, metric_name, title, is_currency=False):
        subset = df[df['Metric'] == metric_name]
        values = [subset['Project'].values[0], subset['Control'].values[0]]
        labels = ['Project', 'Control']

        bars = ax.bar(labels, values, color=colors, width=0.6)
        ax.set_title(title, fontsize=14, pad=15)

        # Add data labels directly on bars
        for bar in bars:
            yval = bar.get_height()
            label_text = f"Tk. {int(yval):,}" if is_currency else f"{yval}%"
            ax.text(bar.get_x() + bar.get_width()/2, yval + (yval*0.02), label_text,
                    ha='center', va='bottom', fontsize=12, fontweight='bold')

    # Plotting each quadrant
    plot_bar(axes[0, 0], 'Monthly Income (BDT)', 'Average Monthly Household Income', is_currency=True)
    plot_bar(axes[0, 1], 'MDD-W Target Met (%)', 'Min. Dietary Diversity for Women (MDD-W)')
    plot_bar(axes[1, 0], 'GAP Adoption (%)', 'Modern Management (GAP) Adoption')
    plot_bar(axes[1, 1], 'Vaccine Use (%)', 'Utilization of Vaccines')

    # Adjust layout
    plt.tight_layout(rect=[0, 0.03, 1, 0.92])
    plt.savefig('impact_dashboard.png', dpi=300, bbox_inches='tight')
    plt.show()

create_impact_dashboard()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

def create_margin_chart():
    # Aesthetic setup
    sns.set_theme(style="white", rc={"axes.spines.top": False, "axes.spines.right": False, "axes.spines.bottom": False})

    # Data extraction from text (11.1, 11.3, 12.1)
    data = pd.DataFrame({
        'Enterprise Type': ['Raw Milk Production', 'Processed Goods (Chhana/Ghee/Meat)', 'Beef Fattening'],
        'Profit Margin (%)': [22, 33, 36]
    }).sort_values(by='Profit Margin (%)', ascending=True)

    fig, ax = plt.subplots(figsize=(10, 5))

    # Create horizontal bar chart
    bars = ax.barh(data['Enterprise Type'], data['Profit Margin (%)'], color='#264653', height=0.5)

    # Add data labels
    for bar in bars:
        width = bar.get_width()
        ax.text(width + 0.5, bar.get_y() + bar.get_height()/2, f"{int(width)}%",
                ha='left', va='center', fontsize=12, fontweight='bold', color='#264653')

    # Styling
    ax.set_title('Average Profit Margins by Enterprise Category', fontsize=16, fontweight='bold', pad=20, loc='left')
    ax.set_xlabel('Profit Margin (%)', fontsize=12, labelpad=10)
    ax.set_ylabel('')
    ax.set_xticks([]) # Remove x-ticks for a cleaner look since we have data labels

    plt.tight_layout()
    plt.savefig('profit_margins.png', dpi=300, bbox_inches='tight')
    plt.show()

create_margin_chart()
```

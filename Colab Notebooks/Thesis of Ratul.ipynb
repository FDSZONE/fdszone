```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Set advanced aesthetic style
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.titleweight'] = 'bold'

# Custom Colors
gold_color = '#D4AF37'
silver_color = '#C0C0C0'
dark_blue = '#1f497d'
mixed_color = '#8C92AC'

def create_income_distribution():
    data = {'Income Range (BDT)': ['Below 20k', '20k–40k', '40k–60k', '60k–100k', 'Above 100k'],
            'Percentage': [15, 30, 25, 20, 10]}
    df = pd.DataFrame(data)

    plt.figure(figsize=(8, 5))
    ax = sns.barplot(x='Income Range (BDT)', y='Percentage', data=df, color=dark_blue, alpha=0.8)
    plt.title('Figure 4.2.1: Income Distribution of Respondents')
    plt.ylabel('Percentage of Respondents (%)')

    # Add data labels
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}%', (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 5), textcoords='offset points')

    sns.despine(left=True)
    plt.tight_layout()
    plt.savefig('fig_4_2_1_income.png', dpi=300)
    plt.show()

def create_investment_preferences():
    labels = ['Gold', 'Both (Gold & Silver)', 'Silver', 'Others']
    sizes = [50, 30, 15, 5]
    colors = [gold_color, mixed_color, silver_color, '#e0e0e0']
    explode = (0.05, 0, 0, 0)  # slightly explode the Gold slice

    plt.figure(figsize=(7, 7))
    plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
            shadow=False, startangle=90, textprops={'fontsize': 12, 'weight': 'bold'})

    # Draw a circle at the center to turn it into a modern donut chart
    centre_circle = plt.Circle((0,0),0.70,fc='white')
    fig = plt.gcf()
    fig.gca().add_artist(centre_circle)

    plt.title('Figure 4.3.1: Overall Investment Preferences')
    plt.tight_layout()
    plt.savefig('fig_4_3_1_preferences.png', dpi=300)
    plt.show()

def create_influencing_factors():
    data = {'Factors': ['Financial Knowledge', 'Market Trends', 'Family Influence', 'Media', 'Friends'],
            'Percentage': [30, 25, 20, 15, 10]}
    df = pd.DataFrame(data).sort_values(by='Percentage', ascending=True)

    plt.figure(figsize=(8, 5))
    ax = sns.barplot(x='Percentage', y='Factors', data=df, hue='Factors', palette='viridis', legend=False)
    plt.title('Figure 4.4.1: Factors Influencing Investment Decisions')
    plt.xlabel('Percentage of Respondents (%)')
    plt.ylabel('')

    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.savefig('fig_4_4_1_factors.png', dpi=300)
    plt.show()

# Execute functions to generate images
create_income_distribution()
create_investment_preferences()
create_influencing_factors()
```

```python
from graphviz import Digraph

def create_thesis_flowchart():
    # Create an advanced directed graph
    dot = Digraph(comment='Middle-Class Investment Decision Model', format='png')
    dot.attr(rankdir='TB', size='8,8', fontname='Helvetica', nodesep='0.5', ranksep='0.8')

    # Define Node Styles
    dot.attr('node', shape='box', style='filled, rounded', fontname='Helvetica', fontsize='12', fontcolor='white')

    # Top Level: Income & Savings
    dot.node('A', 'Middle-Class Income Generation\n(20k - 60k BDT Average)', fillcolor='#1f497d')
    dot.node('B', 'Savings Allocation\n(10% - 20% of Income)', fillcolor='#2b5c8f')

    # Mid Level: The Decision Split (Rational vs Behavioral)
    dot.attr('node', fillcolor='#4a7c59') # Greenish for Rational
    dot.node('C1', 'Rational Factors:\n• Financial Literacy\n• Market Trends')

    dot.attr('node', fillcolor='#c14953') # Reddish for Behavioral
    dot.node('C2', 'Behavioral Factors:\n• Risk Aversion\n• Cultural Values\n• Herd Behavior')

    # Asset Choice Level
    dot.attr('node', fontcolor='black')
    dot.node('D1', 'Preference: GOLD\n(Low Risk, High Stability)', fillcolor='#D4AF37', shape='ellipse')
    dot.node('D2', 'Preference: SILVER\n(Higher Risk, Higher Potential)', fillcolor='#C0C0C0', shape='ellipse')

    # Outcomes
    dot.attr('node', fillcolor='#f4f4f4', fontcolor='black', shape='box')
    dot.node('E', 'Outcome: Financial Stability &\nLong-term Wealth Accumulation')
    dot.node('F', 'Macroeconomic Impact:\nCapital becomes "Dead Capital" unless\nintegrated via Sovereign Gold Bonds')

    # Define Edges (Flow)
    dot.edge('A', 'B')
    dot.edge('B', 'C1')
    dot.edge('B', 'C2')

    dot.edge('C1', 'D1', label=' Primary Driver', fontsize='10')
    dot.edge('C2', 'D1', label=' Cultural & Risk Bias', fontsize='10')

    dot.edge('C1', 'D2', label=' Yield Seeking', fontsize='10')

    dot.edge('D1', 'E')
    dot.edge('D2', 'E')
    dot.edge('E', 'F', style='dashed') # Dashed line for macroeconomic consequence

    # Render and save
    dot.render('fig_5_1_decision_model', view=True)

create_thesis_flowchart()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_conceptual_framework():
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.axis('off')

    # Define colors (Aesthetic palette)
    color_indep = '#2A9D8F'  # Teal
    color_decision = '#E9C46A' # Gold/Yellow
    color_dep = '#E76F51'    # Terracotta
    text_color = 'white'

    # Function to draw rounded boxes
    def draw_box(ax, x, y, width, height, text, bg_color, font_color='white'):
        box = patches.FancyBboxPatch((x, y), width, height, boxstyle="round,pad=0.1,rounding_size=0.1",
                                     edgecolor='none', facecolor=bg_color)
        ax.add_patch(box)
        ax.text(x + width/2, y + height/2, text, ha='center', va='center',
                color=font_color, fontsize=11, fontweight='bold', wrap=True)

    # Independent Variables (Left)
    draw_box(ax, 0, 0.7, 2.5, 0.4, "Socio-Economic\nFactors\n(Income, Culture)", color_indep)
    draw_box(ax, 0, 0.1, 2.5, 0.4, "Psychological &\nBehavioral Biases\n(Herd mentality)", color_indep)

    # Core Decision (Center)
    draw_box(ax, 4.5, 0.4, 2.5, 0.6, "Investment Decision\n(Gold vs. Silver)", color_decision, font_color='black')

    # Dependent Variables (Right)
    draw_box(ax, 9, 0.4, 2.5, 0.6, "Financial Stability\n& Wealth Accumulation", color_dep)

    # Arrows
    arrow_props = dict(facecolor='gray', edgecolor='gray', arrowstyle='-|>', lw=2, mutation_scale=15)

    # Arrows from Independent to Decision
    ax.annotate('', xy=(4.5, 0.7), xytext=(2.6, 0.9), arrowprops=arrow_props)
    ax.annotate('', xy=(4.5, 0.7), xytext=(2.6, 0.3), arrowprops=arrow_props)

    # Arrow from Decision to Dependent
    ax.annotate('', xy=(9, 0.7), xytext=(7.1, 0.7), arrowprops=arrow_props)

    plt.title("Conceptual Framework of the Study", fontsize=16, fontweight='bold', pad=20, color='#264653')
    plt.tight_layout()
    plt.savefig('conceptual_framework.png', dpi=300, bbox_inches='tight')
    plt.show()

draw_conceptual_framework()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

def draw_comparison_chart():
    # Setting up the aesthetic style
    sns.set_theme(style="whitegrid", context="talk")

    # Mock data based on literature review in Chapter 1
    data = {
        'Attribute': ['Historical Stability', 'Affordability', 'Industrial Utility', 'Cultural Value', 'Volatility'],
        'Gold': [9.5, 3.0, 2.0, 9.0, 4.0],
        'Silver': [6.0, 8.5, 8.0, 5.0, 7.5]
    }
    df = pd.DataFrame(data)
    df_melted = df.melt(id_vars='Attribute', var_name='Metal', value_name='Perceived Score (1-10)')

    # Initialize the matplotlib figure
    f, ax = plt.subplots(figsize=(10, 6))

    # Plot the grouped bar chart
    sns.barplot(
        x="Attribute",
        y="Perceived Score (1-10)",
        hue="Metal",
        data=df_melted,
        palette=["#D4AF37", "#C0C0C0"], # Accurate Gold and Silver hex colors
        edgecolor=".2"
    )

    # Customize titles and labels
    ax.set_title("Perceived Investment Attributes: Gold vs. Silver", fontweight='bold', fontsize=16, pad=15)
    ax.set_xlabel("Investment Attribute", fontweight='bold', fontsize=12)
    ax.set_ylabel("Perceived Value / Intensity (1-10)", fontweight='bold', fontsize=12)
    ax.legend(title="Asset", loc='upper left')

    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.savefig('gold_vs_silver_attributes.png', dpi=300, bbox_inches='tight')
    plt.show()

draw_comparison_chart()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_thesis_structure():
    fig, ax = plt.subplots(figsize=(8, 10))
    ax.axis('off')

    # Chapters based on section 1.7
    chapters = [
        ("Chapter 1: Introduction", "Background, Objectives, & Scope"),
        ("Chapter 2: Literature Review", "Prior Studies & Behavioral Finance"),
        ("Chapter 3: Methodology", "Research Design & Data Collection"),
        ("Chapter 4: Data Analysis", "Descriptive Stats & Findings"),
        ("Chapter 5: Discussion", "Interpretation & Theoretical Link"),
        ("Chapter 6: Conclusion", "Summary, Recommendations & Limits")
    ]

    # Aesthetics
    box_width = 4
    box_height = 0.8
    x_center = 2
    y_start = 9
    y_gap = 1.4

    box_color = '#264653' # Deep navy/teal
    text_color = 'white'
    sub_color = '#E9ECEF' # Light gray for subtext

    for i, (title, subtitle) in enumerate(chapters):
        y_pos = y_start - (i * y_gap)

        # Draw Box
        box = patches.FancyBboxPatch(
            (x_center - box_width/2, y_pos), box_width, box_height,
            boxstyle="round,pad=0.2,rounding_size=0.2",
            edgecolor='#2A9D8F', linewidth=2, facecolor=box_color
        )
        ax.add_patch(box)

        # Add Title Text
        ax.text(x_center, y_pos + 0.55, title, ha='center', va='center',
                color=text_color, fontsize=12, fontweight='bold')
        # Add Subtitle Text
        ax.text(x_center, y_pos + 0.25, subtitle, ha='center', va='center',
                color=sub_color, fontsize=10, style='italic')

        # Draw Arrow to next box (except for the last one)
        if i < len(chapters) - 1:
            arrow_props = dict(facecolor='#E76F51', edgecolor='#E76F51',
                               arrowstyle='-|>', lw=2.5, mutation_scale=20)
            ax.annotate('', xy=(x_center, y_pos - 0.2), xytext=(x_center, y_pos),
                        arrowprops=arrow_props)

    plt.title("Thesis Structure Organization", fontsize=16, fontweight='bold', color='#264653')
    plt.tight_layout()
    plt.savefig('thesis_structure.png', dpi=300, bbox_inches='tight')
    plt.show()

draw_thesis_structure()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Set global aesthetic style
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Segoe UI', 'Helvetica', 'Arial', 'sans-serif']
plt.rcParams['text.color'] = '#333333'
plt.rcParams['axes.labelcolor'] = '#333333'
plt.rcParams['xtick.color'] = '#555555'
plt.rcParams['ytick.color'] = '#555555'
plt.rcParams['figure.facecolor'] = '#F8F9FA'
plt.rcParams['axes.facecolor'] = '#FFFFFF'

def draw_flowchart():
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.axis('off')
    ax.set_title("Conceptual Framework of Investment Behavior", fontsize=18, fontweight='bold', pad=20, color='#1A365D')

    # Define box properties
    box_style = "round,pad=0.8"

    # Node definitions (x, y, text, color)
    nodes = {
        'Traditional': (0.2, 0.8, "Economic Determinants\n(Income, Financial Literacy)", "#4299E1"),
        'Psychological': (0.8, 0.8, "Behavioral Biases\n(Loss Aversion, Overconfidence)", "#ED8936"),
        'MiddleClass': (0.5, 0.6, "Middle-Class Investment\nDecision Process", "#9F7AEA"),
        'Gold': (0.3, 0.3, "Gold Investment\n(Low Risk, High Cultural Value)", "#ECC94B"),
        'Silver': (0.7, 0.3, "Silver Investment\n(High Risk, Industrial Utility)", "#A0AEC0"),
        'Outcome': (0.5, 0.1, "Portfolio Formation & Asset Hoarding", "#48BB78")
    }

    # Draw boxes
    for key, (x, y, text, color) in nodes.items():
        bbox_props = dict(boxstyle=box_style, fc=color, ec="white", lw=2, alpha=0.9)
        # Add shadow
        ax.text(x + 0.008, y - 0.008, text, ha="center", va="center", size=11,
                bbox=dict(boxstyle=box_style, fc='gray', ec='none', alpha=0.2))
        # Add actual box
        ax.text(x, y, text, ha="center", va="center", size=12, fontweight='bold', color='white', bbox=bbox_props)

    # Define arrows
    arrow_props = dict(facecolor='#718096', edgecolor='none', width=2.5, headwidth=10, headlength=12, shrink=0.08)

    # Draw connections
    ax.annotate("", xy=(0.45, 0.65), xytext=(0.25, 0.75), arrowprops=arrow_props)
    ax.annotate("", xy=(0.55, 0.65), xytext=(0.75, 0.75), arrowprops=arrow_props)
    ax.annotate("", xy=(0.35, 0.35), xytext=(0.45, 0.55), arrowprops=arrow_props)
    ax.annotate("", xy=(0.65, 0.35), xytext=(0.55, 0.55), arrowprops=arrow_props)
    ax.annotate("", xy=(0.48, 0.15), xytext=(0.35, 0.25), arrowprops=arrow_props)
    ax.annotate("", xy=(0.52, 0.15), xytext=(0.65, 0.25), arrowprops=arrow_props)

    plt.tight_layout()
    plt.savefig('Figure_1_Flowchart.png', dpi=300, bbox_inches='tight')
    plt.show()

def draw_bar_graph():
    fig, ax = plt.subplots(figsize=(10, 6))

    # Categories and mock scores (0 to 10 scale) based on literature
    categories = ['Risk Level', 'Return Volatility', 'Cultural Significance', 'Industrial Utility']
    gold_scores = [3, 4, 9, 2]
    silver_scores = [7, 8, 3, 8]

    x = np.arange(len(categories))
    width = 0.35

    # Aesthetic colors for Gold and Silver
    color_gold = '#D4AF37'
    color_silver = '#8A8D91'

    # Draw bars with slight shadow effect
    ax.bar(x - width/2 + 0.02, gold_scores, width, color='black', alpha=0.1)
    ax.bar(x + width/2 + 0.02, silver_scores, width, color='black', alpha=0.1)

    rects1 = ax.bar(x - width/2, gold_scores, width, label='Gold', color=color_gold, edgecolor='white', linewidth=1.5)
    rects2 = ax.bar(x + width/2, silver_scores, width, label='Silver', color=color_silver, edgecolor='white', linewidth=1.5)

    # Customization
    ax.set_ylabel('Relative Impact Score (0-10)', fontsize=12, fontweight='bold')
    ax.set_title('Comparative Analysis: Gold vs. Silver Investment Profiles', fontsize=16, fontweight='bold', pad=15, color='#1A365D')
    ax.set_xticks(x)
    ax.set_xticklabels(categories, fontsize=11, fontweight='bold')
    ax.legend(fontsize=12, loc='upper left', frameon=True, facecolor='white', edgecolor='#E2E8F0')
    ax.set_ylim(0, 10.5)

    # Value labels on top of bars
    ax.bar_label(rects1, padding=3, fontsize=10, fontweight='bold', color='#555555')
    ax.bar_label(rects2, padding=3, fontsize=10, fontweight='bold', color='#555555')

    plt.tight_layout()
    plt.savefig('Figure_2_BarGraph.png', dpi=300, bbox_inches='tight')
    plt.show()

def draw_donut_chart():
    fig, ax = plt.subplots(figsize=(9, 9))

    # Data derived from theoretical weights in literature
    labels = ['Socio-Cultural Aspects\n(Status, Traditions)',
              'Psychological Biases\n(Risk Perception, Emotions)',
              'Macroeconomic Variables\n(Inflation, Stability)',
              'Financial Literacy\n(Market Knowledge)']
    sizes = [30, 25, 25, 20]
    colors = ['#F6AD55', '#FC8181', '#63B3ED', '#68D391']
    explode = (0.05, 0.05, 0.05, 0.05)

    # Create the donut chart
    wedges, texts, autotexts = ax.pie(sizes, explode=explode, labels=labels, colors=colors,
                                      autopct='%1.1f%%', pctdistance=0.85, shadow=True, startangle=140,
                                      textprops=dict(color="black", fontsize=11, fontweight='bold'))

    # White circle in the middle to make it a donut
    centre_circle = plt.Circle((0, 0), 0.70, fc='#FFFFFF')
    fig.gca().add_artist(centre_circle)

    # Style the percentage text
    for autotext in autotexts:
        autotext.set_color('white')
        autotext.set_fontsize(12)
        autotext.set_fontweight('bold')
        # Add a subtle dark outline to text for readability
        autotext.set_path_effects([plt.matplotlib.patheffects.withStroke(linewidth=2, foreground='black')])

    ax.text(0, 0, 'Decision\nDrivers', ha='center', va='center', fontsize=18, fontweight='bold', color='#2D3748')
    plt.title('Factors Influencing Middle-Class Investment Decisions', fontsize=16, fontweight='bold', pad=20, color='#1A365D')

    plt.tight_layout()
    plt.savefig('Figure_3_DonutChart.png', dpi=300, bbox_inches='tight')
    plt.show()

# Execute generation
if __name__ == "__main__":
    print("Generating Figure 1: Flowchart...")
    draw_flowchart()
    print("Generating Figure 2: Bar Graph...")
    draw_bar_graph()
    print("Generating Figure 3: Donut Chart...")
    draw_donut_chart()
    print("All figures successfully generated and saved as PNG files.")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import seaborn as sns

# ---------------------------------------------------------
# Global Aesthetic Settings for Publication-Ready Figures
# ---------------------------------------------------------
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['Arial', 'Helvetica', 'DejaVu Sans'],
    'axes.titlesize': 16,
    'axes.titleweight': 'bold',
    'axes.labelsize': 12,
    'axes.labelweight': 'bold',
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'figure.dpi': 300,
    'savefig.bbox': 'tight'
})

def generate_figure_1_growth_chart():
    """Generates an aesthetic area/line chart for MFS vs Internet Banking Growth."""
    years = np.arange(2018, 2025)

    # Data derived from Chapter 4 metrics (in Lakh Crore BDT)
    mfs_volume = np.array([3.2, 4.5, 6.8, 9.5, 13.5, 15.2, 17.37])
    ib_volume = np.array([1.2, 1.5, 2.0, 2.8, 3.5, 4.1, 4.8])

    fig, ax = plt.subplots(figsize=(10, 6))

    # Plotting with fill_between for an advanced, modern look
    ax.plot(years, mfs_volume, color='#E2136E', linewidth=3, marker='o', label='MFS Transaction Volume')
    ax.fill_between(years, mfs_volume, alpha=0.15, color='#E2136E')

    ax.plot(years, ib_volume, color='#00529B', linewidth=3, marker='s', label='Internet Banking Volume')
    ax.fill_between(years, ib_volume, alpha=0.15, color='#00529B')

    ax.set_title('Figure 4.1: Transaction Value Growth (2018-2024)', pad=20)
    ax.set_xlabel('Year')
    ax.set_ylabel('Transaction Volume (Lakh Crore BDT)')

    # Remove top and right spines for a cleaner aesthetic
    sns.despine(left=True, bottom=True)
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    ax.legend(loc='upper left', frameon=True, shadow=True)

    # Annotate the final data points
    ax.text(2024, 17.37 + 0.5, '17.37T', color='#E2136E', fontweight='bold', ha='center')
    ax.text(2024, 4.8 + 0.5, '4.8T', color='#00529B', fontweight='bold', ha='center')

    plt.savefig('Figure_4_1_Growth_Trajectory.png')
    plt.close()

def generate_figure_2_market_share():
    """Generates an aesthetic donut chart using actual brand colors."""
    labels = ['bKash', 'Nagad', 'Rocket', 'Others (Upay, etc.)']
    sizes = [63.5, 20.0, 9.0, 7.5]

    # Authentic brand colors for high-end corporate aesthetic
    colors = ['#E2136E', '#F47820', '#8C3494', '#B0B0B0']
    explode = (0.05, 0, 0, 0)  # Explode the dominant player

    fig, ax = plt.subplots(figsize=(8, 8))
    wedges, texts, autotexts = ax.pie(
        sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', startangle=140, pctdistance=0.85,
        wedgeprops=dict(width=0.3, edgecolor='white', linewidth=2),
        textprops=dict(color='black', fontsize=12, weight='bold')
    )

    # Customize the inner percentages for better readability
    for autotext in autotexts:
        autotext.set_color('white')
        autotext.set_fontsize(11)

    ax.set_title('Figure 4.2: MFS Market Share (2024-2025)', pad=20)

    plt.savefig('Figure_4_2_Market_Share.png')
    plt.close()

def generate_figure_3_urban_rural_bar():
    """Generates a grouped bar chart for urban vs rural divides."""
    categories = ['Internet\nPenetration', 'Smartphone\nOwnership', 'Internet Banking\nUser Base']
    urban_stats = [75.0, 71.5, 71.0]
    rural_stats = [35.0, 31.5, 29.0]

    x = np.arange(len(categories))
    width = 0.35

    fig, ax = plt.subplots(figsize=(10, 6))

    # Modern teal and coral contrast
    rects1 = ax.bar(x - width/2, urban_stats, width, label='Urban (%)', color='#008080', edgecolor='none')
    rects2 = ax.bar(x + width/2, rural_stats, width, label='Rural (%)', color='#FF7F50', edgecolor='none')

    ax.set_ylabel('Percentage (%)')
    ax.set_title('Figure 4.3: Urban vs. Rural Digital Divide (2025)', pad=20)
    ax.set_xticks(x)
    ax.set_xticklabels(categories)
    ax.legend(frameon=True, shadow=True)
    sns.despine(left=True, bottom=True)
    ax.grid(axis='y', linestyle='--', alpha=0.7)

    # Add data labels on top of bars
    ax.bar_label(rects1, padding=3, fmt='%.1f%%', weight='bold', color='#008080')
    ax.bar_label(rects2, padding=3, fmt='%.1f%%', weight='bold', color='#FF7F50')

    plt.savefig('Figure_4_3_Urban_Rural_Divide.png')
    plt.close()

def generate_figure_4_flow_chart():
    """Builds a custom regulatory flow chart using matplotlib patches."""
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.axis('off')

    # Define flowchart node properties
    box_style = "round,pad=0.5"

    nodes = {
        'applicant': {'text': 'Applicant\n(Existing Bank / Fintech)', 'xy': (0.5, 0.9), 'color': '#E0F2F1', 'edge': '#00796B'},
        'bb': {'text': 'Bangladesh Bank (BB)\nEvaluation & Due Diligence', 'xy': (0.5, 0.7), 'color': '#E8EAF6', 'edge': '#3F51B5'},
        'prelim': {'text': 'Stage 1: Preliminary Approval\n(Minimum Capital: BDT 400-500 Cr)', 'xy': (0.25, 0.45), 'color': '#FFF3E0', 'edge': '#F57C00'},
        'sandbox': {'text': 'Stage 2: Sandbox Testing\n(12-18 Months Duration)\n• AI Credit Scoring\n• Open Banking APIs', 'xy': (0.25, 0.2), 'color': '#FCE4EC', 'edge': '#C2185B'},
        'license': {'text': 'Stage 3: Full Branchless\nDigital Bank License', 'xy': (0.6, 0.05), 'color': '#E8F5E9', 'edge': '#388E3C'},
        'oversight': {'text': 'Continuous Oversight\n& Compliance:\n\n• e-KYC / Biometrics\n• BFIU (AML/CFT)\n• Risk-Based Supervision', 'xy': (0.75, 0.325), 'color': '#ECEFF1', 'edge': '#455A64'}
    }

    # Draw nodes
    drawn_boxes = {}
    for key, data in nodes.items():
        bbox_props = dict(boxstyle=box_style, facecolor=data['color'], edgecolor=data['edge'], linewidth=2)
        box = ax.text(data['xy'][0], data['xy'][1], data['text'], ha="center", va="center",
                      size=11, weight='bold', bbox=bbox_props, zorder=3)
        drawn_boxes[key] = data['xy']

    # Arrow drawing utility
    def draw_arrow(start, end, connection_style="arc3"):
        ax.annotate('', xy=end, xytext=start,
                    arrowprops=dict(arrowstyle="->", color='#424242', lw=2,
                                    shrinkA=25, shrinkB=25, connectionstyle=connection_style),
                    zorder=2)

    # Draw connections
    draw_arrow(drawn_boxes['applicant'], drawn_boxes['bb'])
    draw_arrow(drawn_boxes['bb'], drawn_boxes['prelim'], "angle3,angleA=0,angleB=-90")
    draw_arrow(drawn_boxes['bb'], drawn_boxes['oversight'], "angle3,angleA=0,angleB=-90")
    draw_arrow(drawn_boxes['prelim'], drawn_boxes['sandbox'])
    draw_arrow(drawn_boxes['sandbox'], drawn_boxes['license'], "angle3,angleA=0,angleB=90")
    draw_arrow(drawn_boxes['oversight'], drawn_boxes['license'], "angle3,angleA=0,angleB=90")

    ax.set_title('Figure 4.4: Digital Bank Licensing Framework (Guidelines V2, 2025)', pad=20, fontsize=16, weight='bold')

    plt.savefig('Figure_4_4_Regulatory_Flowchart.png')
    plt.close()

# Execute all functions to generate images
generate_figure_1_growth_chart()
generate_figure_2_market_share()
generate_figure_3_urban_rural_bar()
generate_figure_4_flow_chart()

print("All advanced visual figures successfully generated as PNG files.")
```
